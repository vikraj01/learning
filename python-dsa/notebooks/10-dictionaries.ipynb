# 10. Dictionaries

After this notebook you'll be able to store "this goes with that" pairs, look things up by name, change them, and loop over them.

Dictionaries are in a huge number of LeetCode answers, so this notebook and the next one go slowly.

## The contacts app on your phone

Your phone's contacts app doesn't store numbers in a numbered row. You don't think "Mum is contact number 37". You type **Mum**, and the phone finds her number straight away.

Each entry is a pair: a **name**, and the **number** that goes with it. The name is how you look things up.

Python has the same thing. It's called a **dictionary** (`dict` for short). You'll also hear **map** or **hash map**. All three mean the same thing here.

- The thing you look up by (the name) is the **key**.
- The thing you get back (the number) is the **value**.
- A key and its value together are a **key-value pair**.

A list is a shelf where you find things by **position** (0, 1, 2...). A dictionary is a contacts app where you find things by **name**.

## Making one and reading from it

Curly brackets `{ }`. Each pair is written `key: value`, with commas between the pairs. Run this:

````python
contacts = {"Mum": "98100 11111", "Vikash": "98200 22222"}
print(contacts["Mum"])
print(len(contacts))
print(contacts)
````

You'll see:

```
98100 11111
2
{'Mum': '98100 11111', 'Vikash': '98200 22222'}
```

- `contacts["Mum"]` looks up the key `"Mum"` and gives you its value. Same square brackets as a list, but you put a **key** inside, not a position.
- `len` counts the **pairs**, not the keys and values separately. Two pairs, so `2`.

The lookup is fast, however big the dictionary gets. Your phone doesn't read every contact to find Mum, and Python doesn't either. That speed is why dictionaries matter so much on LeetCode.

Keys are usually strings or numbers. Values can be anything: numbers, strings, even lists.

## A key that isn't there

What if you ask for a name that isn't saved? Run this. It breaks on purpose.

````python
contacts = {"Mum": "98100 11111", "Vikash": "98200 22222"}
print(contacts["Riya"])
````

The last line of the error says:

```
KeyError: 'Riya'
```

A **KeyError** means "there's no key called `'Riya'` in this dictionary". Python doesn't guess; it stops.

### `get`: the safe way to ask

**`get`** asks politely. If the key is there, you get its value. If it isn't, you get `None` (Python's word for "nothing"), or a **default** value you choose:

````python
print(contacts.get("Mum"))
print(contacts.get("Riya"))
print(contacts.get("Riya", "not saved"))
print(contacts.get("Riya", 0))
````

```
98100 11111
None
not saved
0
```

- `contacts.get("Riya")` gives `None`, with no error.
- `contacts.get("Riya", "not saved")` gives your default instead.
- Remember `.get(key, 0)`. In the next notebook it's the heart of counting.

`get` only **reads**. It doesn't add Riya to the dictionary.

## Adding and changing

Put a value in with `=`. The same line does two different jobs:

- If the key is **new**, the pair is **added**.
- If the key is **already there**, its value is **replaced**.

````python
contacts = {"Mum": "98100 11111", "Vikash": "98200 22222"}

contacts["Riya"] = "98300 33333"     # new key: added
print(contacts)

contacts["Mum"] = "98100 99999"      # old key: replaced
print(contacts)
print(len(contacts))
````

```
{'Mum': '98100 11111', 'Vikash': '98200 22222', 'Riya': '98300 33333'}
{'Mum': '98100 99999', 'Vikash': '98200 22222', 'Riya': '98300 33333'}
3
```

A dictionary can't hold the same key twice. Like your phone, there's one entry per name. Changing Mum's number didn't make a second Mum; the length stayed `3`.

New pairs go at the end. A dictionary remembers the order you added things in.

### Changing a value using its old value

Your marks went up by 5 after a re-check. Read the old value, add to it, store it back:

````python
marks = {"maths": 70, "science": 82}
marks["maths"] = marks["maths"] + 5
marks["science"] += 3        # same idea, shorter
print(marks)
````

```
{'maths': 75, 'science': 85}
```

Read `marks["maths"] = marks["maths"] + 5` from the right: look up `70`, add `5`, store `75` back under `"maths"`.

`+=` works too, but **only if the key is already there**. You'll see why that matters in the next notebook.

### Removing a pair

`del` removes a key and its value. `pop` removes it and gives you the value back:

````python
marks = {"maths": 75, "science": 85, "art": 90}
del marks["art"]
print(marks)

science = marks.pop("science")
print(science)
print(marks)
````

```
{'maths': 75, 'science': 85}
85
{'maths': 75}
```

## `in` checks the keys, not the values

This one catches everyone at least once. **Predict** each line before you run it.

````python
contacts = {"Mum": "98100 11111", "Riya": "98300 33333"}
print("Riya" in contacts)
print("Aman" in contacts)
print("98300 33333" in contacts)
print("98300 33333" in contacts.values())
````

```
True
False
False
True
```

- `"Riya" in contacts` is `True`: `"Riya"` is a key.
- `"98300 33333" in contacts` is `False`, even though that number is right there. It's a **value**, and `in` only looks at **keys**.
- To search the values, ask `in contacts.values()`.

`in` on keys is fast. `in .values()` has to look through every value, so it's slow on big dictionaries. Most of the time you'll design your dictionary so that the thing you search for is the key.

`in` works well with `if`, to avoid a KeyError:

````python
name = "Aman"
if name in contacts:
    print(contacts[name])
else:
    print(name, "is not saved")
````

```
Aman is not saved
```

## Looping over a dictionary

There are three ways, and they give you different things. Run each one and look closely.

### 1. A plain `for`: you get the keys

````python
menu = {"tea": 10, "coffee": 30, "samosa": 15}
for item in menu:
    print(item)
````

```
tea
coffee
samosa
```

Only the keys. To get the price, look it up inside the loop:

````python
for item in menu:
    print(item, "costs", menu[item])
````

```
tea costs 10
coffee costs 30
samosa costs 15
```

Step by step:

| Round | `item` | `menu[item]` | printed |
| --- | --- | --- | --- |
| 1 | `"tea"` | `10` | `tea costs 10` |
| 2 | `"coffee"` | `30` | `coffee costs 30` |
| 3 | `"samosa"` | `15` | `samosa costs 15` |

### 2. `.items()`: key and value together

**`.items()`** hands you each pair. Give two names in the `for`, like with `enumerate`: the first gets the key, the second gets the value.

````python
for item, price in menu.items():
    print(item, "->", price)
````

```
tea -> 10
coffee -> 30
samosa -> 15
```

This is the one you'll use most. It reads nicely: "for each item and its price".

### 3. `.values()`: only the values

Useful when you only care about the numbers, for example the total or the biggest:

````python
print(sum(menu.values()))
print(max(menu.values()))
for price in menu.values():
    print(price)
````

```
55
30
10
30
15
```

`max(menu.values())` gives the biggest **price** (`30`), not the name of the item. To find the **name** of the most expensive item, loop with `.items()` and remember the best one so far (you'll do this in a task).

### A loop with an `if` inside

Which items cost less than 20? Predict, then run.

````python
cheap = []
for item, price in menu.items():
    if price < 20:
        cheap.append(item)
print(cheap)
````

```
['tea', 'samosa']
```

| Round | `item` | `price` | `price < 20`? | `cheap` afterwards |
| --- | --- | --- | --- | --- |
| 1 | `"tea"` | `10` | yes | `['tea']` |
| 2 | `"coffee"` | `30` | no | `['tea']` |
| 3 | `"samosa"` | `15` | yes | `['tea', 'samosa']` |

## A dictionary as a lookup table

A very common use: the dictionary is a **price list** (or a score list, or a translation table), and you look things up in it while you loop over something else.

Your order at the tea stall is a list. The menu is a dictionary. What's the bill?

````python
menu = {"tea": 10, "coffee": 30, "samosa": 15}
order = ["tea", "samosa", "samosa", "tea"]

total = 0
for item in order:
    total += menu[item]
    print("added", item, "-> total is now", total)
print("bill:", total)
````

```
added tea -> total is now 10
added samosa -> total is now 25
added samosa -> total is now 40
added tea -> total is now 50
bill: 50
```

The loop walks through the **order** (the list). The dictionary only answers "how much is this?" each time round.

**Try it:** add `"coffee"` to the order and run it again. What's the new bill?

## Common mistakes

**Using a position, like a list.** Dictionaries have no position 0. Run this:

````python
menu = {"tea": 10, "coffee": 30}
print(menu[0])
````

```
KeyError: 0
```

Python looked for a **key** called `0`. There isn't one. Use a key: `menu["tea"]`.

**Forgetting the quotes around a key.** `menu[tea]` (no quotes) means "the variable called `tea`", which doesn't exist. You'll get `NameError: name 'tea' is not defined`. String keys need quotes.

**Expecting `in` to search the values.** It only looks at keys. Use `x in d.values()` for values.

**Expecting `max(d)` to give the biggest value.** `max(menu)` compares the **keys** (the item names, alphabetically). Use `max(menu.values())` for the biggest value.
