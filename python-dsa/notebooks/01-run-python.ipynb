## How a notebook works

This page is a **notebook**. It's a page where you can read and run code in the same place.

A notebook is made of boxes called **cells**. There are two kinds:

- **Text cells**, like this one. You only read them.
- **Code cells**, the grey boxes with `[ ]:` on the left. They hold Python code you can run.

To **run** a code cell:

1. Click on it.
2. Press **Shift + Enter** (hold Shift, press Enter).

The answer appears right under the cell, and the cursor moves to the next cell. A number shows up in the brackets, like `[1]:`. It counts the order you ran cells in.

Two more things to know:

- Your notebook saves itself every few minutes. **Ctrl + S** saves it now.
- If things ever look strange, use the menu **Kernel → Restart Kernel and Clear Outputs of All Cells**. It's like switching Python off and on again. Then run the cells again from the top.

### A program is a recipe

A recipe is a list of steps: boil water, add tea, add milk, wait two minutes. You follow the steps from top to bottom, one at a time.

A **program** is the same thing for a computer: a list of steps, written down, that the computer follows from top to bottom. **Python** is the language we write those steps in. It looks a bit like English, which makes it a good first language.

The computer is a very careful cook, though. It does exactly what the recipe says, nothing more. If a word is misspelled, it stops and tells you. That's normal, and you'll see it happen later in this notebook.

## `print` shows things on the screen

Run the cell below.

````python
print("Hello")
print(2 + 3)
print("2 + 3")
````

You'll see:

```

```

- **`print(...)`** shows whatever is inside the brackets.

- With quotes, `"2 + 3"` is only text, so it prints `2 + 3`.

## Errors are normal

Everyone who writes code sees errors every day. An error isn't you breaking something. It's Python saying "I didn't understand this line, here's where and why".

The cell below has a typo on purpose: `prnt` instead of `print`. Run it.

````python
print("Hello")
prnt("Bye")
````

You'll see `Hello`, and then a block of red text, something like:

```
NameError                                 Traceback (most recent call last)
Cell In[6], line 2
      1 print("Hello")
----> 2 prnt("Bye")

NameError: name 'prnt' is not defined
```

That block is called a **traceback**. It looks scary, but read it like this:

1. **Read the last line first.** It says what went wrong: `NameError: name 'prnt' is not defined`. Python doesn't know any word called `prnt`.
2. **Then find the arrow.** `---->` points at the line that broke, and `line 2` says which line it is.
3. Notice that `Hello` was printed first. Python followed the recipe until it hit the step it didn't understand, then stopped.

**Fix it:** click the cell, change `prnt` to `print`, and run it again. Now both lines print.

## Common mistake: a missing bracket

Every `(` needs a `)`, and every `"` needs another `"`. If one is missing, Python can't even read the line. Run this:

````python
print("one"
````

You'll see an error ending with:

```
SyntaxError: '(' was never closed
```

A **SyntaxError** means Python couldn't read the line, because something is missing or out of place. Here, a `(` was opened and never closed. Add the `)` at the end and run it again.
