## 1. The shopping bill

**How to think about it:** Two jobs in one walk over the basket: add up what we know, and collect what we don't. A set collects the unknown names without repeats, and `sorted` turns it into a list at the end.

```python
def bill(prices, basket):
    total = 0
    unknown = set()
    for item in basket:
        if item in prices:
            total += prices[item]
        else:
            unknown.add(item)
    return total, sorted(unknown)
```

```python
prices = {"milk": 30, "bread": 25, "eggs": 60}
check(bill(prices, ["milk", "bread", "milk", "jam"]), (85, ["jam"]))
check(bill(prices, ["tea", "jam", "tea"]), (0, ["jam", "tea"]))
check(bill(prices, []), (0, []))
check(bill(prices, ["eggs"]), (60, []))
check(bill(prices, ["eggs", "eggs", "eggs", "milk"]), (210, []))
check(bill({}, ["milk"]), (0, ["milk"]))
```

## 2. The library's visitor book

**How to think about it:** Clean first (lowercase), then count with a dictionary. Everything we need comes out of that one dictionary: its size is the number of people, and the names with a count of 2+ are the regulars.

```python
def visitors(names):
    counts = {}
    for name in names:
        name = name.lower()
        counts[name] = counts.get(name, 0) + 1

    regulars = []
    for name, times in counts.items():
        if times >= 2:
            regulars.append(name)

    return len(names), len(counts), sorted(regulars)
```

```python
check(visitors(["Asha", "ravi", "asha", "Meena", "Ravi", "asha"]), (6, 3, ["asha", "ravi"]))
check(visitors(["Tom"]), (1, 1, []))
check(visitors([]), (0, 0, []))
check(visitors(["a", "b", "c"]), (3, 3, []))
check(visitors(["ZARA", "zara", "Zara"]), (3, 1, ["zara"]))
check(visitors(["x", "Y", "y", "X", "z"]), (5, 3, ["x", "y"]))
```

## 3. Turn the dictionary inside out

**How to think about it:** Each `(key, value)` pair adds one name to one list. The "is it there yet? if not, start an empty list" step is the same idea as counting, with a list instead of a number. Sort the lists at the end.

```python
def invert(d):
    result = {}
    for key, value in d.items():
        if value not in result:
            result[value] = []
        result[value].append(key)

    for value in result:
        result[value].sort()
    return result
```

```python
check(invert({"apple": "red", "cherry": "red", "banana": "yellow"}), {"red": ["apple", "cherry"], "yellow": ["banana"]})
check(invert({"Asha": 90, "Ravi": 75, "Meena": 90}), {90: ["Asha", "Meena"], 75: ["Ravi"]})
check(invert({}), {})
check(invert({"x": 1}), {1: ["x"]})
check(invert({"c": 0, "b": 0, "a": 0}), {0: ["a", "b", "c"]})

colours = {"sky": "blue", "sea": "blue"}
invert(colours)
check(colours, {"sky": "blue", "sea": "blue"})
```

## 4. The most used word

**How to think about it:** Clean, count, find the top. The "top" step has two passes: first find the highest count, then collect every word with that count. That second pass is what handles ties.

```python
def top_words(text):
    counts = {}
    for word in text.split():
        word = word.lower().strip(".,!?")
        if word == "":
            continue
        counts[word] = counts.get(word, 0) + 1

    if len(counts) == 0:
        return []

    best = max(counts.values())
    winners = []
    for word, times in counts.items():
        if times == best:
            winners.append(word)
    return sorted(winners)
```

```python
check(top_words("The cat and the dog. THE end!"), ["the"])
check(top_words("red blue Red BLUE green"), ["blue", "red"])
check(top_words(""), [])
check(top_words("!! ?? ..."), [])
check(top_words("Tea? tea, TEA. coffee!"), ["tea"])
check(top_words("one"), ["one"])
check(top_words("b a c"), ["a", "b", "c"])
```

## 5. Anagram groups

**How to think about it:** The trick is a key that is the same for all anagrams: the letters in sorted order. Then it's the "group by key" pattern from problem 3: label → list of words.

```python
def group_anagrams(words):
    groups = {}
    for word in words:
        label = "".join(sorted(word))
        if label not in groups:
            groups[label] = []
        groups[label].append(word)

    result = []
    for group in groups.values():
        result.append(sorted(group))
    return sorted(result)
```

```python
check(group_anagrams(["listen", "silent", "cat", "enlist", "act", "dog"]), [["act", "cat"], ["dog"], ["enlist", "listen", "silent"]])
check(group_anagrams(["ab", "ba", "ab"]), [["ab", "ab", "ba"]])
check(group_anagrams([]), [])
check(group_anagrams(["tea"]), [["tea"]])
check(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]), [["ate", "eat", "tea"], ["bat"], ["nat", "tan"]])
check(group_anagrams(["a", "b", "a"]), [["a", "a"], ["b"]])
```

## 6. Two shops, one stock list

**How to think about it:** Combining two count dictionaries is the counting pattern again, but adding a number instead of 1. The trap is building on top of `shop_a`: start from an empty dictionary so the inputs stay as they were.

```python
def restock(shop_a, shop_b, minimum):
    total = {}
    for item, count in shop_a.items():
        total[item] = total.get(item, 0) + count
    for item, count in shop_b.items():
        total[item] = total.get(item, 0) + count

    low = []
    for item, count in total.items():
        if count < minimum:
            low.append(item)
    return total, sorted(low)
```

```python
check(restock({"pens": 10, "paper": 3}, {"pens": 2, "glue": 1, "paper": 4}, 5), ({"pens": 12, "paper": 7, "glue": 1}, ["glue"]))
check(restock({"ink": 0}, {}, 1), ({"ink": 0}, ["ink"]))
check(restock({}, {}, 5), ({}, []))
check(restock({"a": 1}, {"b": 1}, 2), ({"a": 1, "b": 1}, ["a", "b"]))
check(restock({"tape": 5}, {"tape": 5}, 10), ({"tape": 10}, []))

a = {"pens": 10}
b = {"pens": 5}
restock(a, b, 3)
check(a, {"pens": 10})
check(b, {"pens": 5})
```

## 7. The class gradebook

**How to think about it:** 7a is two counting dictionaries side by side (sum and count). 7b is the "best so far" pattern from notebook 09, kept per subject in a dictionary, with one extra rule for ties.

```python
def averages(records):
    totals = {}
    counts = {}
    for name, subject, mark in records:
        totals[name] = totals.get(name, 0) + mark
        counts[name] = counts.get(name, 0) + 1

    result = {}
    for name in totals:
        result[name] = round(totals[name] / counts[name], 1)
    return result


def best_in_each_subject(records):
    best = {}  # subject -> (mark, name)
    for name, subject, mark in records:
        if subject not in best:
            best[subject] = (mark, name)
        else:
            best_mark, best_name = best[subject]
            if mark > best_mark or (mark == best_mark and name < best_name):
                best[subject] = (mark, name)

    result = {}
    for subject, (mark, name) in best.items():
        result[subject] = name
    return result
```

```python
records = [
    ("Asha", "maths", 90),
    ("Ravi", "maths", 72),
    ("Asha", "art", 65),
    ("Meena", "art", 88),
    ("Ravi", "art", 88),
    ("Meena", "maths", 81),
]
check(averages(records), {"Asha": 77.5, "Ravi": 80.0, "Meena": 84.5})
check(best_in_each_subject(records), {"maths": "Asha", "art": "Meena"})
check(averages([]), {})
check(best_in_each_subject([]), {})
check(averages([("Tom", "music", 50)]), {"Tom": 50.0})
check(best_in_each_subject([("Tom", "music", 50)]), {"music": "Tom"})
check(averages([("Ana", "x", 70), ("Ana", "y", 80), ("Ana", "z", 82)]), {"Ana": 77.3})
check(best_in_each_subject([("Zoe", "pe", 60), ("Bo", "pe", 60), ("Al", "pe", 59)]), {"pe": "Bo"})
```

## 8. Does the sentence follow the pattern?

**How to think about it:** One dictionary only stops a letter from changing its word. The second dictionary, going the other way, stops two letters from sharing a word. Both directions together make the matching one-to-one.

```python
def follows_pattern(pattern, sentence):
    words = sentence.split()
    if len(words) != len(pattern):
        return False

    letter_to_word = {}
    word_to_letter = {}
    for i in range(len(pattern)):
        letter = pattern[i]
        word = words[i]
        if letter in letter_to_word and letter_to_word[letter] != word:
            return False
        if word in word_to_letter and word_to_letter[word] != letter:
            return False
        letter_to_word[letter] = word
        word_to_letter[word] = letter
    return True
```

```python
check(follows_pattern("abba", "dog cat cat dog"), True)
check(follows_pattern("abba", "dog cat cat fish"), False)
check(follows_pattern("abba", "dog dog dog dog"), False)
check(follows_pattern("abc", "dog cat"), False)
check(follows_pattern("a", "tea"), True)
check(follows_pattern("aaa", "go go go"), True)
check(follows_pattern("ab", "go go"), False)
check(follows_pattern("abc", "red blue green"), True)
```

## 9. The first repeat

**How to think about it:** Returning the moment we meet an item the second time automatically gives "the earliest second appearance". The `seen` dictionary remembers where each item was first, so we can report both positions.

```python
def first_repeat(items):
    seen = {}  # item -> index where we first saw it
    for i, item in enumerate(items):
        if item in seen:
            return item, seen[item], i
        seen[item] = i
    return None
```

```python
check(first_repeat([3, 8, 5, 8, 3]), (8, 1, 3))
check(first_repeat(list("banana")), ("a", 1, 3))
check(first_repeat([1, 2, 3]), None)
check(first_repeat([]), None)
check(first_repeat([7, 7]), (7, 0, 1))
check(first_repeat(["tea", "milk", "sugar", "milk", "tea", "tea"]), ("milk", 1, 3))
```

## 10. Have I been here before?

**How to think about it:** A point is two numbers that belong together, so it's a tuple, and tuples can live in a set. Then it's "seen before?" again: check, then add. The start must be in the set from the beginning.

```python
def first_revisit(path):
    x, y = 0, 0
    visited = {(0, 0)}
    for step in path:
        step = step.upper()
        if step == "N":
            y += 1
        elif step == "S":
            y -= 1
        elif step == "E":
            x += 1
        elif step == "W":
            x -= 1
        point = (x, y)
        if point in visited:
            return point
        visited.add(point)
    return None
```

```python
check(first_revisit("NNESWS"), (0, 1))
check(first_revisit("NS"), (0, 0))
check(first_revisit("NNEESS"), None)
check(first_revisit(""), None)
check(first_revisit("nnesws"), (0, 1))
check(first_revisit("EEWW"), (1, 0))
check(first_revisit("ENWS"), (0, 0))
```

## 11. ⭐ Stretch: the class election

**How to think about it:** The tie rule needs two kinds of knowledge: the final top score (only known at the end) and the order things happened (only seen during counting). Two passes give us both: first find the target, then replay the count and stop at the first person to hit it.

```python
def election(ballots):
    if len(ballots) == 0:
        return ""

    # Pass 1: the final counts, and the top number of votes.
    counts = {}
    for name in ballots:
        counts[name] = counts.get(name, 0) + 1
    target = max(counts.values())

    # Pass 2: replay the count; the first to reach the target wins.
    running = {}
    for name in ballots:
        running[name] = running.get(name, 0) + 1
        if running[name] == target:
            return name
```

```python
check(election(["Asha", "Ravi", "Ravi", "Asha", "Meena"]), "Ravi")
check(election(["a", "b", "b", "a", "a"]), "a")
check(election(["x", "y", "z"]), "x")
check(election([]), "")
check(election(["Meena"]), "Meena")
check(election(["b", "a", "a", "b", "c", "c"]), "a")
check(election(["p", "q", "q", "p", "p", "q"]), "p")
```

## 12. ⭐ Stretch: people you may know

**How to think about it:** Counting again: every time a candidate shows up through one of our friends, that's one more mutual friend. The ordering is the tricky part, and tuples solve it: `(-mutual, name)` sorts by most mutual first, then by name.

```python
def suggest_friends(friends, person):
    if person not in friends:
        return []

    my_friends = set(friends[person])
    mutual = {}
    for friend in my_friends:
        for candidate in friends.get(friend, []):
            if candidate != person and candidate not in my_friends:
                mutual[candidate] = mutual.get(candidate, 0) + 1

    # Negative counts so the biggest comes first; ties go alphabetical.
    order = []
    for name, count in mutual.items():
        order.append((-count, name))

    result = []
    for negative_count, name in sorted(order):
        result.append((name, -negative_count))
    return result
```

```python
friends = {
    "asha": ["ravi", "meena"],
    "ravi": ["asha", "meena", "tom", "zara"],
    "meena": ["asha", "ravi", "tom"],
    "tom": ["ravi", "meena"],
    "zara": ["ravi"],
    "lily": [],
}
check(suggest_friends(friends, "asha"), [("tom", 2), ("zara", 1)])
check(suggest_friends(friends, "zara"), [("asha", 1), ("meena", 1), ("tom", 1)])
check(suggest_friends(friends, "lily"), [])
check(suggest_friends(friends, "nobody"), [])
check(suggest_friends(friends, "tom"), [("asha", 2), ("zara", 1)])
check(suggest_friends(friends, "ravi"), [])
```
