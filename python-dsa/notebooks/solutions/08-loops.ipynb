### Task 2: add up 1 to n

The total box goes **before** the loop. `range(1, n + 1)` reaches `n`. For `n = 0` the range is empty, so the loop never runs and `0` comes back.

```python
def sum_up_to(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

check(sum_up_to(4), 10)
check(sum_up_to(1), 1)
check(sum_up_to(100), 5050)
check(sum_up_to(0), 0)
```

### Task 3: add up the digits

```python
def digit_sum(n):
    total = 0
    while n > 0:
        total += n % 10
        n = n // 10
    return total

check(digit_sum(472), 13)
check(digit_sum(5), 5)
check(digit_sum(1001), 2)
check(digit_sum(0), 0)
```

### Task 4: where's the first negative?

Start with `-1` (the answer when there's none). The `break` makes sure a later negative doesn't replace the first one.

```python
def first_negative(nums):
    answer = -1
    for i, n in enumerate(nums):
        if n < 0:
            answer = i
            break
    return answer

check(first_negative([4, 7, -2, 5, -9]), 2)
check(first_negative([-1, 3]), 0)
check(first_negative([1, 2, 3]), -1)
check(first_negative([]), -1)
```

### Task 5: Final Value of Variable After Performing Operations

```python
class Solution:
    def finalValueAfterOperations(self, operations: list[str]) -> int:
        x = 0
        for op in operations:
            if "+" in op:
                x += 1
            else:
                x -= 1
        return x

check(Solution().finalValueAfterOperations(["--X", "X++", "X++"]), 1)
check(Solution().finalValueAfterOperations(["++X", "++X", "X++"]), 3)
check(Solution().finalValueAfterOperations(["X++", "++X", "--X", "X--"]), 0)
```

### Task 6: Number of Steps to Reduce a Number to Zero

A `while` loop, because the number of steps isn't known in advance.

```python
class Solution:
    def numberOfSteps(self, num: int) -> int:
        steps = 0
        while num > 0:
            if num % 2 == 0:
                num = num // 2
            else:
                num -= 1
            steps += 1
        return steps

check(Solution().numberOfSteps(14), 6)
check(Solution().numberOfSteps(8), 4)
check(Solution().numberOfSteps(123), 12)
check(Solution().numberOfSteps(0), 0)
```

### Task 7: Fizz Buzz

The "3 and 5" check must come first. If `% 3` came first, 15 would stop there and become `"Fizz"`.

```python
class Solution:
    def fizzBuzz(self, n: int) -> list[str]:
        answer = []
        for i in range(1, n + 1):
            if i % 3 == 0 and i % 5 == 0:
                answer.append("FizzBuzz")
            elif i % 3 == 0:
                answer.append("Fizz")
            elif i % 5 == 0:
                answer.append("Buzz")
            else:
                answer.append(str(i))
        return answer

check(Solution().fizzBuzz(3), ["1", "2", "Fizz"])
check(Solution().fizzBuzz(5), ["1", "2", "Fizz", "4", "Buzz"])
check(Solution().fizzBuzz(15), ["1", "2", "Fizz", "4", "Buzz", "Fizz", "7", "8", "Fizz", "Buzz", "11", "Fizz", "13", "14", "FizzBuzz"])
check(Solution().fizzBuzz(1), ["1"])
```
