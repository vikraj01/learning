### Task 1: 2235. Add Two Integers

```python
class Solution:
    def sum(self, num1: int, num2: int) -> int:
        return num1 + num2

check(Solution().sum(12, 5), 17)
check(Solution().sum(-10, 4), -6)
check(Solution().sum(-3, -7), -10)
```

### Task 2: 2651. Calculate Delayed Arrival Time

`% 24` takes away whole days and keeps the hour.

```python
class Solution:
    def findDelayedArrivalTime(self, arrivalTime: int, delayedTime: int) -> int:
        return (arrivalTime + delayedTime) % 24

check(Solution().findDelayedArrivalTime(15, 5), 20)
check(Solution().findDelayedArrivalTime(13, 11), 0)
check(Solution().findDelayedArrivalTime(23, 3), 2)
```

### Task 3: 2413. Smallest Even Multiple

If `n` is even, it's already the answer. If it's odd, double it.

```python
class Solution:
    def smallestEvenMultiple(self, n: int) -> int:
        if n % 2 == 0:
            return n
        return n * 2

check(Solution().smallestEvenMultiple(5), 10)
check(Solution().smallestEvenMultiple(6), 6)
check(Solution().smallestEvenMultiple(1), 2)
```

### Task 4: 2481. Minimum Cuts to Divide a Circle

One slice needs no cuts. An even number of slices uses cuts across (each makes 2 edges), so `n // 2`. An odd number needs `n` cuts from the middle.

```python
class Solution:
    def numberOfCuts(self, n: int) -> int:
        if n == 1:
            return 0
        if n % 2 == 0:
            return n // 2
        return n

check(Solution().numberOfCuts(4), 2)
check(Solution().numberOfCuts(3), 3)
check(Solution().numberOfCuts(1), 0)
check(Solution().numberOfCuts(6), 3)
```

### Task 5: 2469. Convert the Temperature

```python
class Solution:
    def convertTemperature(self, celsius: float) -> list[float]:
        kelvin = celsius + 273.15
        fahrenheit = celsius * 1.80 + 32.00
        return [kelvin, fahrenheit]

check(Solution().convertTemperature(36.5), [309.65, 97.7])
check(Solution().convertTemperature(0), [273.15, 32.0])
check(Solution().convertTemperature(122.11), [395.26, 251.798])
```
