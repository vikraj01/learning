### Task 1: place in the alphabet

Position from 0, plus 1.

```python
def alphabet_place(ch):
    return ord(ch) - ord("a") + 1

check(alphabet_place("a"), 1)
check(alphabet_place("c"), 3)
check(alphabet_place("z"), 26)
```

### Task 2: move a letter along

```python
def shift_letter(ch, k):
    pos = ord(ch) - ord("a")
    new_pos = (pos + k) % 26
    return chr(new_pos + ord("a"))

check(shift_letter("a", 2), "c")
check(shift_letter("y", 3), "b")
check(shift_letter("z", 1), "a")
check(shift_letter("m", 0), "m")
```

### Task 3: add up the digits

For `0` the loop never runs, so the total stays 0.

```python
def digit_sum(n):
    total = 0
    while n > 0:
        total += n % 10
        n = n // 10
    return total

check(digit_sum(4072), 13)
check(digit_sum(7), 7)
check(digit_sum(0), 0)
```

### Task 4: longest word first

`sorted` keeps ties in their old order, even with `reverse=True`.

```python
def sort_by_length(words):
    return sorted(words, key=len, reverse=True)

check(sort_by_length(["kiwi", "banana", "fig", "mango"]), ["banana", "mango", "kiwi", "fig"])
check(sort_by_length(["bb", "a", "cc"]), ["bb", "cc", "a"])
check(sort_by_length([]), [])
```

### Task 5: top of the class

`max` gives the whole pair; `[0]` takes the name. On a tie, `max` keeps the first one.

```python
def top_student(pairs):
    best = max(pairs, key=lambda p: p[1])
    return best[0]

check(top_student([("Tanvi", 82), ("Riya", 95), ("Aman", 78)]), "Riya")
check(top_student([("Tanvi", 90), ("Aman", 90)]), "Tanvi")
check(top_student([("Neha", 55)]), "Neha")
```

### Task 6: Check if the Sentence Is Pangram

```python
class Solution:
    def checkIfPangram(self, sentence: str) -> bool:
        counts = [0] * 26
        for ch in sentence:
            counts[ord(ch) - ord("a")] += 1
        return all(c > 0 for c in counts)

check(Solution().checkIfPangram("thequickbrownfoxjumpsoverthelazydog"), True)
check(Solution().checkIfPangram("leetcode"), False)
check(Solution().checkIfPangram("abcdefghijklmnopqrstuvwxyz"), True)
```

A shorter way, with a set: `return len(set(sentence)) == 26`.

### Task 7: Sort the People

```python
class Solution:
    def sortPeople(self, names: list[str], heights: list[int]) -> list[str]:
        pairs = list(zip(names, heights))
        pairs.sort(key=lambda p: p[1], reverse=True)
        return [name for name, height in pairs]

check(Solution().sortPeople(["Mary", "John", "Emma"], [180, 165, 170]), ["Mary", "Emma", "John"])
check(Solution().sortPeople(["Alice", "Bob", "Bob"], [155, 185, 150]), ["Bob", "Alice", "Bob"])
check(Solution().sortPeople(["Tanvi"], [160]), ["Tanvi"])
```
