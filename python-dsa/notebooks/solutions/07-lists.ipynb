### Task 1: ends of the shelf

```python
def ends_sum(nums):
    return nums[0] + nums[-1]

check(ends_sum([4, 8, 15]), 19)
check(ends_sum([1, 2, 3, 4]), 5)
check(ends_sum([7]), 14)
```

### Task 2: add it twice, in place

`append` changes the shelf itself, so the caller sees it. No `return` needed.

```python
def add_twice(cart, item):
    cart.append(item)
    cart.append(item)

cart = ["milk"]
add_twice(cart, "tea")
check(cart, ["milk", "tea", "tea"])

empty = []
add_twice(empty, "sugar")
check(empty, ["sugar", "sugar"])
```

### Task 3: top two marks

`sorted` makes a new list, so `marks` stays as it was. Then a slice takes the first two.

```python
def top_two(marks):
    return sorted(marks, reverse=True)[:2]

marks = [70, 45, 90, 60]
check(top_two(marks), [90, 70])
check(marks, [70, 45, 90, 60])
check(top_two([5, 5, 1]), [5, 5])
```

### Task 4: Concatenation of Array

```python
class Solution:
    def getConcatenation(self, nums: list[int]) -> list[int]:
        return nums + nums

check(Solution().getConcatenation([1, 2, 1]), [1, 2, 1, 1, 2, 1])
check(Solution().getConcatenation([1, 3, 2, 1]), [1, 3, 2, 1, 1, 3, 2, 1])
```

### Task 5: Reverse String

`reverse()` flips the list itself. `s = s[::-1]` would only move the name inside the function.

```python
class Solution:
    def reverseString(self, s: list[str]) -> None:
        s.reverse()

s = ["h", "e", "l", "l", "o"]
Solution().reverseString(s)
check(s, ["o", "l", "l", "e", "h"])

s = ["H", "a", "n", "n", "a", "h"]
Solution().reverseString(s)
check(s, ["h", "a", "n", "n", "a", "H"])
```

### Task 6: Maximum Product of Two Elements

After sorting, the two biggest are the last two. (`sorted` keeps LeetCode's list as it was, which is polite.)

```python
class Solution:
    def maxProduct(self, nums: list[int]) -> int:
        ordered = sorted(nums)
        return (ordered[-1] - 1) * (ordered[-2] - 1)

check(Solution().maxProduct([3, 4, 5, 2]), 12)
check(Solution().maxProduct([1, 5, 4, 5]), 16)
check(Solution().maxProduct([3, 7]), 12)
```
