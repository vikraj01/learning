### Task 2: a rectangle

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

check(Rectangle(3, 4).area(), 12)
check(Rectangle(3, 4).perimeter(), 14)
check(Rectangle(5, 5).area(), 25)
check(Rectangle(1, 10).perimeter(), 22)
```

### Task 4: a piggy bank

The money lives on `self.money`, so every method sees the same amount.

```python
class PiggyBank:
    def __init__(self):
        self.money = 0

    def add(self, amount):
        self.money += amount

    def spend(self, amount):
        if amount > self.money:
            return False
        self.money -= amount
        return True

    def total(self):
        return self.money

bank = PiggyBank()
check(bank.total(), 0)
bank.add(10)
bank.add(5)
check(bank.total(), 15)
check(bank.spend(20), False)
check(bank.total(), 15)
check(bank.spend(12), True)
check(bank.total(), 3)
```

### Task 5: two counters, kept apart

`red` and `blue` are separate objects, so each has its own `self.clicks`.

```python
class ClickCounter:
    def __init__(self):
        self.clicks = 0

    def click(self):
        self.clicks += 1

    def count(self):
        return self.clicks

    def reset(self):
        self.clicks = 0

red = ClickCounter()
blue = ClickCounter()
red.click()
red.click()
red.click()
blue.click()
check(red.count(), 3)
check(blue.count(), 1)
red.reset()
check(red.count(), 0)
check(blue.count(), 1)
```

### Task 6: LeetCode 1603, Design Parking System

A list of free spaces, where `self.free[1]` is big, `[2]` medium, `[3]` small. Index 0 is never used; it's there so `carType` can be the index directly.

```python
class ParkingSystem:
    def __init__(self, big: int, medium: int, small: int):
        self.free = [0, big, medium, small]

    def addCar(self, carType: int) -> bool:
        if self.free[carType] == 0:
            return False
        self.free[carType] -= 1
        return True

p = ParkingSystem(1, 1, 0)
check(p.addCar(1), True)
check(p.addCar(2), True)
check(p.addCar(3), False)
check(p.addCar(1), False)

p2 = ParkingSystem(0, 0, 2)
check(p2.addCar(3), True)
check(p2.addCar(3), True)
check(p2.addCar(3), False)
```

### Task 7: LeetCode 933, Number of Recent Calls

Keep every time, then count the ones from `t - 3000` to `t`. (Notebook 21 shows a faster way with a `deque`, which throws old calls away.)

```python
class RecentCounter:
    def __init__(self):
        self.times = []

    def ping(self, t: int) -> int:
        self.times.append(t)
        count = 0
        for time in self.times:
            if time >= t - 3000:
                count += 1
        return count

rc = RecentCounter()
check(rc.ping(1), 1)
check(rc.ping(100), 2)
check(rc.ping(3001), 3)
check(rc.ping(3002), 3)
check(rc.ping(7000), 1)
```
