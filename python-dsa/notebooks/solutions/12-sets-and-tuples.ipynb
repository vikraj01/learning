### Task 1: how many different?

```python
def count_different(nums):
    return len(set(nums))

check(count_different([3, 1, 3, 2, 1]), 3)
check(count_different([5, 5, 5]), 1)
check(count_different([]), 0)
```

### Task 2: lowest and highest

Start both at the first mark, then walk through the rest. `return low, high` returns a tuple.

```python
def lowest_and_highest(marks):
    low = marks[0]
    high = marks[0]
    for m in marks:
        if m < low:
            low = m
        if m > high:
            high = m
    return low, high

check(lowest_and_highest([67, 92, 45, 80]), (45, 92))
check(lowest_and_highest([50]), (50, 50))
check(lowest_and_highest([3, 3, 1]), (1, 3))
```

### Task 3 (LeetCode 217): Contains Duplicate

If the set is shorter than the list, something was repeated. (The `seen` set loop works too.)

```python
class Solution:
    def containsDuplicate(self, nums):
        return len(set(nums)) < len(nums)

check(Solution().containsDuplicate([1, 2, 3, 1]), True)
check(Solution().containsDuplicate([1, 2, 3, 4]), False)
check(Solution().containsDuplicate([1, 1, 1, 3, 3, 4, 3, 2, 4, 2]), True)
check(Solution().containsDuplicate([7]), False)
```

### Task 4 (LeetCode 771): Jewels and Stones

Put the jewels in a set, so each `in` check is fast.

```python
class Solution:
    def numJewelsInStones(self, jewels, stones):
        jewel_set = set(jewels)
        count = 0
        for s in stones:
            if s in jewel_set:
                count += 1
        return count

check(Solution().numJewelsInStones("aA", "aAAbbbb"), 3)
check(Solution().numJewelsInStones("z", "ZZ"), 0)
check(Solution().numJewelsInStones("abc", ""), 0)
```

### Task 5 (LeetCode 1832): Check if the Sentence Is Pangram

All 26 letters are there exactly when the set of letters has 26 items.

```python
class Solution:
    def checkIfPangram(self, sentence):
        return len(set(sentence)) == 26

check(Solution().checkIfPangram("thequickbrownfoxjumpsoverthelazydog"), True)
check(Solution().checkIfPangram("leetcode"), False)
check(Solution().checkIfPangram("abcdefghijklmnopqrstuvwxy"), False)
```

### Task 6 (LeetCode 349): Intersection of Two Arrays

`&` gives what's in both; `sorted` turns the set into a sorted list.

```python
class Solution:
    def intersection(self, nums1, nums2):
        return sorted(set(nums1) & set(nums2))

check(Solution().intersection([1, 2, 2, 1], [2, 2]), [2])
check(Solution().intersection([4, 9, 5], [9, 4, 9, 8, 4]), [4, 9])
check(Solution().intersection([1, 2], [3, 4]), [])
```

### Task 7 (LeetCode 136): Single Number

Each pair adds then removes its number, so only the single one is left.

```python
class Solution:
    def singleNumber(self, nums):
        s = set()
        for n in nums:
            if n in s:
                s.remove(n)
            else:
                s.add(n)
        return s.pop()

check(Solution().singleNumber([2, 2, 1]), 1)
check(Solution().singleNumber([4, 1, 2, 1, 2]), 4)
check(Solution().singleNumber([1]), 1)
```
