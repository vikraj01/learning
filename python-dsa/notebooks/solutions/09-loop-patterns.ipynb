### Task 1: squares

Build a list. (Short way: `return [n * n for n in nums]`.)

```python
def squares(nums):
    result = []
    for n in nums:
        result.append(n * n)
    return result

check(squares([1, 2, 3]), [1, 4, 9])
check(squares([-2, 0, 5]), [4, 0, 25])
check(squares([]), [])
```

### Task 2: long words

Counter: add 1, not the word.

```python
def count_long(words, k):
    count = 0
    for w in words:
        if len(w) > k:
            count += 1
    return count

check(count_long(["tea", "mango", "kiwi", "banana"], 4), 2)
check(count_long(["a", "bb"], 5), 0)
check(count_long(["hello", "world"], 0), 2)
```

### Task 3: the coldest day

Best so far, flipped to `<`, starting from the first item so negatives work.

```python
def coldest(temps):
    best = temps[0]
    for t in temps:
        if t < best:
            best = t
    return best

check(coldest([12, 5, 9]), 5)
check(coldest([3, 7, 1]), 1)
check(coldest([-5, -2, -8]), -8)
check(coldest([4]), 4)
```

### Task 4: any big spender?

`return True` inside the loop, `return False` after it.

```python
def any_over(prices, limit):
    for p in prices:
        if p > limit:
            return True
    return False

check(any_over([10, 20, 500], 100), True)
check(any_over([500, 10], 100), True)
check(any_over([10, 20, 30], 100), False)
check(any_over([], 100), False)
```

### Task 5: Running Sum of 1d Array

Total + build a list, like the bank balance example.

```python
class Solution:
    def runningSum(self, nums: list[int]) -> list[int]:
        total = 0
        result = []
        for n in nums:
            total += n
            result.append(total)
        return result

check(Solution().runningSum([1, 2, 3, 4]), [1, 3, 6, 10])
check(Solution().runningSum([1, 1, 1, 1, 1]), [1, 2, 3, 4, 5])
check(Solution().runningSum([3, 1, 2, 10, 1]), [3, 4, 6, 16, 17])
```

### Task 6: Richest Customer Wealth

`sum(customer)` is each customer's total; then best so far. Starting at `0` is fine here, because wealth can't be negative.

```python
class Solution:
    def maximumWealth(self, accounts: list[list[int]]) -> int:
        best = 0
        for customer in accounts:
            wealth = sum(customer)
            if wealth > best:
                best = wealth
        return best

check(Solution().maximumWealth([[1, 2, 3], [3, 2, 1]]), 6)
check(Solution().maximumWealth([[1, 5], [7, 3], [3, 5]]), 10)
check(Solution().maximumWealth([[2, 8, 7], [7, 1, 3], [1, 9, 5]]), 17)
```

### Task 7: Number of Good Pairs

Every pair + a counter. `j` starts at `i + 1`, which also makes sure `i < j`.

```python
class Solution:
    def numIdenticalPairs(self, nums: list[int]) -> int:
        count = 0
        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):
                if nums[i] == nums[j]:
                    count += 1
        return count

check(Solution().numIdenticalPairs([1, 2, 3, 1, 1, 3]), 4)
check(Solution().numIdenticalPairs([1, 1, 1, 1]), 6)
check(Solution().numIdenticalPairs([1, 2, 3]), 0)
```

### Task 8: Max Consecutive Ones

The finished trace for `[1, 1, 0, 1, 1, 1]`:

| `n` | current run | best |
| --- | --- | --- |
| 1 | 1 | 1 |
| 1 | 2 | 2 |
| 0 | 0 | 2 |
| 1 | 1 | 2 |
| 1 | 2 | 2 |
| 1 | 3 | 3 |

```python
class Solution:
    def findMaxConsecutiveOnes(self, nums: list[int]) -> int:
        current = 0
        best = 0
        for n in nums:
            if n == 1:
                current += 1
                if current > best:
                    best = current
            else:
                current = 0
        return best

check(Solution().findMaxConsecutiveOnes([1, 1, 0, 1, 1, 1]), 3)
check(Solution().findMaxConsecutiveOnes([1, 0, 1, 1, 0, 1]), 2)
check(Solution().findMaxConsecutiveOnes([0, 0]), 0)
check(Solution().findMaxConsecutiveOnes([1]), 1)
```
