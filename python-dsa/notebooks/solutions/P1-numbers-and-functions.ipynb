## 1. Leap year

The rules have exceptions on top of exceptions, so check them from the most specific (400) to the least (4). The first rule that matches decides the answer.

```python
def is_leap(year):
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False
```

```python
check(is_leap(2024), True)
check(is_leap(2023), False)
check(is_leap(1900), False)
check(is_leap(2000), True)
check(is_leap(2100), False)
check(is_leap(1600), True)
```

## 2. Rock, paper, scissors

Check the bad input first, otherwise `"lizard", "lizard"` looks like a draw. After that there are only three ways for A to win. Any other valid pair that isn't a draw is a win for B.

```python
def is_move(m):
    return m == "rock" or m == "paper" or m == "scissors"


def winner(a, b):
    if not is_move(a) or not is_move(b):
        return "invalid"
    if a == b:
        return "draw"
    if (a == "rock" and b == "scissors") or (a == "scissors" and b == "paper") or (a == "paper" and b == "rock"):
        return "A"
    return "B"
```

```python
check(winner("rock", "scissors"), "A")
check(winner("scissors", "rock"), "B")
check(winner("paper", "rock"), "A")
check(winner("paper", "scissors"), "B")
check(winner("rock", "rock"), "draw")
check(winner("rock", "lizard"), "invalid")
check(winner("Rock", "paper"), "invalid")
check(winner("lizard", "lizard"), "invalid")
```

## 3. Round to the nearest multiple

`n % m` is how far you are above the multiple below. Compare it with half of `m`. Multiplying both sides by 2 (`leftover * 2 >= m`) keeps everything in whole numbers, and `>=` makes halfway go up.

```python
def round_to_nearest(n, m):
    leftover = n % m
    below = n - leftover
    if leftover * 2 >= m:
        return below + m
    return below
```

```python
check(round_to_nearest(23, 5), 25)
check(round_to_nearest(22, 5), 20)
check(round_to_nearest(25, 5), 25)
check(round_to_nearest(14, 4), 16)
check(round_to_nearest(13, 4), 12)
check(round_to_nearest(0, 7), 0)
check(round_to_nearest(3, 10), 0)
check(round_to_nearest(5, 10), 10)
check(round_to_nearest(7, 1), 7)
```

## 4. Seconds to a clock

Each piece is one `//` and one `%`. Hours: whole hours, then `% 24` to drop days. Minutes: what's left after the hours, in whole minutes. Seconds: what's left after the minutes. `clock` shifts hours four places and minutes two places to the left by multiplying.

```python
def hours_of(s):
    return (s // 3600) % 24


def minutes_of(s):
    return (s % 3600) // 60


def seconds_of(s):
    return s % 60


def clock(s):
    return hours_of(s) * 10000 + minutes_of(s) * 100 + seconds_of(s)
```

```python
check(hours_of(3725), 1)
check(minutes_of(3725), 2)
check(seconds_of(3725), 5)
check(clock(3725), 10205)
check(clock(0), 0)
check(clock(59), 59)
check(clock(86399), 235959)
check(clock(86400), 0)
check(clock(90061), 10101)
```

## 5. Three-digit number tricks

Both functions start the same way: refuse anything outside 100 to 999, then split the number into its hundreds, tens and ones with `//` and `%`. After that, each is one line of maths.

```python
def reverse3(n):
    if n < 100 or n > 999:
        return -1
    hundreds = n // 100
    tens = n // 10 % 10
    ones = n % 10
    return ones * 100 + tens * 10 + hundreds


def is_armstrong3(n):
    if n < 100 or n > 999:
        return False
    hundreds = n // 100
    tens = n // 10 % 10
    ones = n % 10
    return hundreds ** 3 + tens ** 3 + ones ** 3 == n
```

```python
check(reverse3(123), 321)
check(reverse3(120), 21)
check(reverse3(100), 1)
check(reverse3(905), 509)
check(reverse3(99), -1)
check(reverse3(1000), -1)
check(reverse3(-123), -1)
check(is_armstrong3(153), True)
check(is_armstrong3(370), True)
check(is_armstrong3(407), True)
check(is_armstrong3(123), False)
check(is_armstrong3(100), False)
check(is_armstrong3(1), False)
```

## 6. What kind of triangle?

Throw out the bad cases first: a side of 0 or less, or a pair of sides that isn't longer than the third. Since the sides come in any order, test all three pairs. Then go from the most equal (all three) to the least.

```python
class Solution:
    def triangleType(self, a: int, b: int, c: int) -> str:
        if a <= 0 or b <= 0 or c <= 0:
            return "invalid"
        if not (a + b > c and a + c > b and b + c > a):
            return "invalid"
        if a == b and b == c:
            return "equilateral"
        if a == b or b == c or a == c:
            return "isosceles"
        return "scalene"
```

```python
check(Solution().triangleType(3, 3, 3), "equilateral")
check(Solution().triangleType(3, 4, 5), "scalene")
check(Solution().triangleType(5, 5, 8), "isosceles")
check(Solution().triangleType(8, 5, 5), "isosceles")
check(Solution().triangleType(5, 8, 5), "isosceles")
check(Solution().triangleType(1, 2, 3), "invalid")
check(Solution().triangleType(1, 1, 5), "invalid")
check(Solution().triangleType(0, 0, 0), "invalid")
check(Solution().triangleType(-3, 4, 5), "invalid")
```

## 7. Parking fee

Three rules, checked in order: free first (it applies to both days), then the weekend flat price, then the weekday price. The tricky bit is "every hour started", which is dividing and rounding **up**: adding 59 before `// 60` does that with whole numbers.

```python
def parking_fee(minutes, is_weekend):
    if minutes <= 30:
        return 0
    if is_weekend:
        return 50
    hours_started = (minutes + 59) // 60
    fee = hours_started * 20
    if fee > 150:
        fee = 150
    return fee
```

```python
check(parking_fee(0, False), 0)
check(parking_fee(30, False), 0)
check(parking_fee(31, False), 20)
check(parking_fee(60, False), 20)
check(parking_fee(61, False), 40)
check(parking_fee(420, False), 140)
check(parking_fee(450, False), 150)
check(parking_fee(600, False), 150)
check(parking_fee(10, True), 0)
check(parking_fee(45, True), 50)
check(parking_fee(600, True), 50)
```

## 8. Electricity bill

Each slab is only charged for the units that fall inside it. Pick the case by where the units end, and add the full price of the slabs below it. The ₹50 fixed charge goes on top in every case.

```python
def bill(units):
    if units < 0:
        return -1
    if units <= 100:
        energy = units * 3
    elif units <= 200:
        energy = 100 * 3 + (units - 100) * 5
    else:
        energy = 100 * 3 + 100 * 5 + (units - 200) * 8
    return energy + 50
```

```python
check(bill(0), 50)
check(bill(1), 53)
check(bill(100), 350)
check(bill(101), 355)
check(bill(150), 600)
check(bill(200), 850)
check(bill(250), 1250)
check(bill(-5), -1)
```

## 9. Is the shop open?

One number is easier to compare than two, so change the time into minutes since midnight. Then each day type is one or two ranges, with `<=` at the start (opening counts) and `<` at the end (closing doesn't). Bad hours or minutes are rejected first.

```python
def is_weekday(day):
    return day == "mon" or day == "tue" or day == "wed" or day == "thu" or day == "fri"


def is_open(day, hour, minute):
    if hour < 0 or hour > 23 or minute < 0 or minute > 59:
        return False
    t = hour * 60 + minute   # minutes since midnight
    if is_weekday(day):
        morning = 9 * 60 <= t < 13 * 60
        afternoon = 14 * 60 <= t < 18 * 60
        return morning or afternoon
    if day == "sat":
        return 10 * 60 <= t < 14 * 60
    return False
```

```python
check(is_open("mon", 9, 0), True)
check(is_open("mon", 8, 59), False)
check(is_open("wed", 13, 0), False)
check(is_open("wed", 13, 59), False)
check(is_open("wed", 14, 0), True)
check(is_open("fri", 17, 59), True)
check(is_open("fri", 18, 0), False)
check(is_open("sat", 13, 30), True)
check(is_open("sat", 14, 0), False)
check(is_open("sun", 11, 0), False)
check(is_open("tue", 25, 0), False)
check(is_open("tue", 10, 60), False)
```

## 10. The change counter

Start with the change still owed and a count of 0. For each piece from biggest to smallest, `//` says how many fit and `%` says what's left. Nine pieces means nine pairs of the same two lines; with a loop later, this becomes three lines.

```python
def change_pieces(price, paid):
    if paid < price:
        return -1
    left = paid - price
    pieces = 0

    pieces = pieces + left // 500
    left = left % 500
    pieces = pieces + left // 200
    left = left % 200
    pieces = pieces + left // 100
    left = left % 100
    pieces = pieces + left // 50
    left = left % 50
    pieces = pieces + left // 20
    left = left % 20
    pieces = pieces + left // 10
    left = left % 10
    pieces = pieces + left // 5
    left = left % 5
    pieces = pieces + left // 2
    left = left % 2
    pieces = pieces + left   # whatever is left is ₹1 coins

    return pieces
```

```python
check(change_pieces(37, 50), 3)
check(change_pieces(363, 500), 5)
check(change_pieces(100, 50), -1)
check(change_pieces(0, 0), 0)
check(change_pieces(250, 250), 0)
check(change_pieces(1, 1000), 9)
check(change_pieces(12, 2000), 11)
```

## 11. ⭐ Stretch: tomorrow's date

Tomorrow is today + 1, then fix whatever overflowed: a day past the end of the month rolls over to day 1 of the next month, and a month past 12 rolls over to January of the next year. `days_in_month` keeps that logic out of `next_day`, and leans on `is_leap` for February.

```python
def is_leap(year):
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False


def days_in_month(month, year):
    if month == 2:
        if is_leap(year):
            return 29
        return 28
    if month == 4 or month == 6 or month == 9 or month == 11:
        return 30
    return 31


def next_day(day, month, year):
    day = day + 1
    if day > days_in_month(month, year):
        day = 1
        month = month + 1
    if month > 12:
        month = 1
        year = year + 1
    return year * 10000 + month * 100 + day
```

```python
check(days_in_month(1, 2023), 31)
check(days_in_month(4, 2023), 30)
check(days_in_month(2, 2023), 28)
check(days_in_month(2, 2024), 29)
check(days_in_month(2, 1900), 28)
check(next_day(15, 6, 2024), 20240616)
check(next_day(28, 2, 2023), 20230301)
check(next_day(28, 2, 2024), 20240229)
check(next_day(29, 2, 2024), 20240301)
check(next_day(30, 4, 2024), 20240501)
check(next_day(31, 1, 2024), 20240201)
check(next_day(31, 12, 2023), 20240101)
```

## 12. ⭐ Stretch: the angle between clock hands

Find where each hand points, measured from 12 o'clock. The difference between them is one of the two angles; if it's over 180°, the other way round (360 minus it) is the smaller one. `% 12` folds the 24-hour clock onto the 12-hour face.

```python
def clock_angle(hour, minute):
    hour_hand = (hour % 12) * 30 + minute * 0.5
    minute_hand = minute * 6
    diff = hour_hand - minute_hand
    if diff < 0:
        diff = -diff
    if diff > 180:
        diff = 360 - diff
    return diff
```

```python
check(clock_angle(3, 0), 90)
check(clock_angle(12, 0), 0)
check(clock_angle(0, 0), 0)
check(clock_angle(6, 0), 180)
check(clock_angle(3, 30), 75)
check(clock_angle(9, 45), 22.5)
check(clock_angle(12, 15), 82.5)
check(clock_angle(13, 0), 30)
check(clock_angle(11, 59), 5.5)
```
