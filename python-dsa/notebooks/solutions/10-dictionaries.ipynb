### Task 1: price with a fallback

`get` with a default of `0` does it in one line.

```python
def get_price(menu, item):
    return menu.get(item, 0)

check(get_price({"tea": 10, "coffee": 30}, "tea"), 10)
check(get_price({"tea": 10, "coffee": 30}, "coffee"), 30)
check(get_price({"tea": 10, "coffee": 30}, "pizza"), 0)
check(get_price({}, "tea"), 0)
```

### Task 2: price goes up

Check with `in` first, so a missing item doesn't cause a KeyError.

```python
def raise_price(menu, item, extra):
    if item in menu:
        menu[item] += extra
    return menu

check(raise_price({"tea": 10, "coffee": 30}, "tea", 2), {"tea": 12, "coffee": 30})
check(raise_price({"tea": 10, "coffee": 30}, "coffee", 5), {"tea": 10, "coffee": 35})
check(raise_price({"tea": 10}, "pizza", 5), {"tea": 10})
```

### Task 3: the bill

Loop over the order; the menu answers "how much?" each time.

```python
def total_bill(menu, order):
    total = 0
    for item in order:
        total += menu[item]
    return total

check(total_bill({"tea": 10, "samosa": 15}, ["tea", "samosa", "tea"]), 35)
check(total_bill({"tea": 10, "coffee": 30}, ["coffee"]), 30)
check(total_bill({"tea": 10}, []), 0)
```

### Task 4: who passed?

```python
def passed(marks):
    names = []
    for name, mark in marks.items():
        if mark >= 40:
            names.append(name)
    return names

check(passed({"Tanvi": 78, "Riya": 35, "Aman": 40}), ["Tanvi", "Aman"])
check(passed({"Riya": 10, "Aman": 39}), [])
check(passed({}), [])
```

### Task 5: is this number saved?

`number in contacts` would look at the names (keys). The numbers are the values.

```python
def is_saved(contacts, number):
    return number in contacts.values()

check(is_saved({"Mum": "111", "Riya": "333"}, "333"), True)
check(is_saved({"Mum": "111", "Riya": "333"}, "Riya"), False)
check(is_saved({"Mum": "111"}, "999"), False)
```

### Task 6: the most expensive item

Start with nothing as the best, and replace it whenever you find a higher price.

```python
def most_expensive(menu):
    best_item = None
    best_price = -1
    for item, price in menu.items():
        if price > best_price:
            best_item = item
            best_price = price
    return best_item

check(most_expensive({"tea": 10, "coffee": 30, "samosa": 15}), "coffee")
check(most_expensive({"cake": 50, "tea": 10}), "cake")
check(most_expensive({"water": 5}), "water")
```

### Task 7 (LeetCode 13): Roman to Integer

For each letter, peek at the next one. `i + 1 < len(s)` makes sure there **is** a next one, so the last letter doesn't cause an IndexError.

```python
class Solution:
    def romanToInt(self, s):
        value = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}
        total = 0
        for i in range(len(s)):
            if i + 1 < len(s) and value[s[i]] < value[s[i + 1]]:
                total -= value[s[i]]
            else:
                total += value[s[i]]
        return total

check(Solution().romanToInt("III"), 3)
check(Solution().romanToInt("IV"), 4)
check(Solution().romanToInt("IX"), 9)
check(Solution().romanToInt("LVIII"), 58)
check(Solution().romanToInt("MCMXCIV"), 1994)
```
