### Task 1: swap the ends

For a list of one item, box 0 and box -1 are the same box, so the swap changes nothing. That's fine.

````python
def swap_ends(nums):
    nums[0], nums[-1] = nums[-1], nums[0]

a = [1, 2, 3, 4]
swap_ends(a)
check(a, [4, 2, 3, 1])

b = ["tea", "milk"]
swap_ends(b)
check(b, ["milk", "tea"])

c = [7]
swap_ends(c)
check(c, [7])
````

### Task 2: palindrome, with two pointers

For `""`, `right` starts at -1, so `left < right` is false straight away and it returns `True`.

````python
def is_palindrome(word):
    left = 0
    right = len(word) - 1
    while left < right:
        if word[left] != word[right]:
            return False
        left += 1
        right -= 1
    return True

check(is_palindrome("level"), True)
check(is_palindrome("noon"), True)
check(is_palindrome("tea"), False)
check(is_palindrome("abca"), False)
check(is_palindrome("a"), True)
check(is_palindrome(""), True)
````

### Task 3: count the swaps

````python
def count_swaps(n):
    left = 0
    right = n - 1
    swaps = 0
    while left < right:
        swaps += 1
        left += 1
        right -= 1
    return swaps

check(count_swaps(5), 2)
check(count_swaps(4), 2)
check(count_swaps(7), 3)
check(count_swaps(1), 0)
check(count_swaps(0), 0)
````

### Task 4: a pair that adds up

````python
def pair_for(prices, budget):
    left = 0
    right = len(prices) - 1
    while left < right:
        total = prices[left] + prices[right]
        if total == budget:
            return [left, right]
        elif total > budget:
            right -= 1
        else:
            left += 1

check(pair_for([10, 20, 35, 50, 75], 70), [1, 3])
check(pair_for([1, 2, 3, 4, 5], 9), [3, 4])
check(pair_for([1, 2, 3, 4, 5], 3), [0, 1])
check(pair_for([5, 5], 10), [0, 1])
````

### Task 5: Reverse String (LeetCode 344)

````python
class Solution:
    def reverseString(self, s: list[str]) -> None:
        left = 0
        right = len(s) - 1
        while left < right:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -= 1

s = ["h", "e", "l", "l", "o"]
Solution().reverseString(s)
check(s, ["o", "l", "l", "e", "h"])

s = ["H", "a", "n", "n", "a", "h"]
Solution().reverseString(s)
check(s, ["h", "a", "n", "n", "a", "H"])

s = ["x"]
Solution().reverseString(s)
check(s, ["x"])
````

### Task 6: Move Zeroes (LeetCode 283)

Keep every number that isn't 0, packed to the front. Then the boxes from `write` to the end get 0.

````python
class Solution:
    def moveZeroes(self, nums: list[int]) -> None:
        write = 0
        for read in range(len(nums)):
            if nums[read] != 0:
                nums[write] = nums[read]
                write += 1
        for i in range(write, len(nums)):
            nums[i] = 0

nums = [0, 1, 0, 3, 12]
Solution().moveZeroes(nums)
check(nums, [1, 3, 12, 0, 0])

nums = [0]
Solution().moveZeroes(nums)
check(nums, [0])

nums = [4, 2, 0, 0, 7]
Solution().moveZeroes(nums)
check(nums, [4, 2, 7, 0, 0])

nums = [1, 2, 3]
Solution().moveZeroes(nums)
check(nums, [1, 2, 3])
````

### Task 7: Remove Duplicates from Sorted Array (LeetCode 26)

Same as `keep_one_of_each` in the notebook.

````python
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        write = 1
        for read in range(1, len(nums)):
            if nums[read] != nums[write - 1]:
                nums[write] = nums[read]
                write += 1
        return write

nums = [1, 1, 2]
k = Solution().removeDuplicates(nums)
check(k, 2)
check(nums[:k], [1, 2])

nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k = Solution().removeDuplicates(nums)
check(k, 5)
check(nums[:k], [0, 1, 2, 3, 4])

nums = [7]
k = Solution().removeDuplicates(nums)
check(k, 1)
check(nums[:k], [7])
````

### Task 8: Squares of a Sorted Array (LeetCode 977)

The first-try way, square and sort, is also correct:

```python
return sorted([n * n for n in nums])
```

The two-pointer way fills `result` from the back, biggest square first. `while left <= right` here (not `<`), because the last remaining number also has to be placed.

````python
class Solution:
    def sortedSquares(self, nums: list[int]) -> list[int]:
        result = [0] * len(nums)
        left = 0
        right = len(nums) - 1
        pos = len(nums) - 1
        while left <= right:
            left_square = nums[left] * nums[left]
            right_square = nums[right] * nums[right]
            if left_square > right_square:
                result[pos] = left_square
                left += 1
            else:
                result[pos] = right_square
                right -= 1
            pos -= 1
        return result

check(Solution().sortedSquares([-4, -1, 0, 3, 10]), [0, 1, 9, 16, 100])
check(Solution().sortedSquares([-7, -3, 2, 3, 11]), [4, 9, 9, 49, 121])
check(Solution().sortedSquares([1, 2, 3]), [1, 4, 9])
check(Solution().sortedSquares([-3, -2, -1]), [1, 4, 9])
check(Solution().sortedSquares([5]), [25])
````
