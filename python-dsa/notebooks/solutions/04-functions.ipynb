### Task 3: is it positive?

A comparison already gives `True` or `False`, so you can return it directly. An `if` with `return True` / `return False` is also right.

```python
def is_positive(n):
    return n > 0

check(is_positive(5), True)
check(is_positive(-2), False)
check(is_positive(0), False)
```

### Task 4: the last digit

```python
def last_digit(n):
    return n % 10

check(last_digit(1234), 4)
check(last_digit(7), 7)
check(last_digit(50), 0)
```

### Task 5: the bigger one

```python
def bigger(a, b):
    if a > b:
        return a
    return b

check(bigger(3, 8), 8)
check(bigger(10, 2), 10)
check(bigger(-1, -5), -1)
check(bigger(4, 4), 4)
```

### Task 6: grades

```python
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 50:
        return "C"
    return "Try again"

check(grade(95), "A")
check(grade(90), "A")
check(grade(89), "B")
check(grade(50), "C")
check(grade(49), "Try again")
```

### Task 7: use another function

```python
def is_even(n):
    if n % 2 == 0:
        return True
    return False

def is_odd_square(n):
    return not is_even(square(n))

check(is_odd_square(3), True)
check(is_odd_square(4), False)
check(is_odd_square(0), False)
```

### Task 8: hours and minutes as one number

135 // 60 is 2 hours, 135 % 60 is 15 minutes, and 2 × 100 + 15 is 215.

```python
def film_length(minutes):
    hours = minutes // 60
    mins = minutes % 60
    return hours * 100 + mins

check(film_length(135), 215)
check(film_length(60), 100)
check(film_length(45), 45)
check(film_length(0), 0)
```
