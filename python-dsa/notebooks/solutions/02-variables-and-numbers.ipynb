### Task 1: the tea bill

### Task 2: a long film

```python
film = 200
hours = film // 60
mins = film % 60
check(hours, 3)
check(mins, 20)
```

### Task 3: sharing sweets

```python
sweets = 29
friends = 4
each = sweets // friends
left = sweets % friends
check(each, 7)
check(left, 1)
```

### Task 4: take a number apart

```python
n = 9876
last = n % 10
rest = n // 10
check(last, 6)
check(rest, 987)
```

### Task 5: a growing score

```python
points = 10
points += 5
points *= 2
check(points, 30)
```
