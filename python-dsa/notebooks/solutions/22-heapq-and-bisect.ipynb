### Task 2: sorting with a heap

`heap = nums[:]` makes a copy first, so `heapify` doesn't change the list you were given.

```python
def heap_sort(nums):
    heap = nums[:]
    heapq.heapify(heap)
    result = []
    while heap:
        result.append(heapq.heappop(heap))
    return result

check(heap_sort([3, 1, 2]), [1, 2, 3])
check(heap_sort([5, 5, 1, 9]), [1, 5, 5, 9])
check(heap_sort([]), [])
```

### Task 3: LeetCode 1046, Last Stone Weight

Every stone goes in with a minus, so the heaviest comes out first. `-heapq.heappop(heap)` puts the minus back.

```python
class Solution:
    def lastStoneWeight(self, stones: list[int]) -> int:
        heap = []
        for s in stones:
            heapq.heappush(heap, -s)

        while len(heap) > 1:
            first = -heapq.heappop(heap)
            second = -heapq.heappop(heap)
            if first != second:
                heapq.heappush(heap, -(first - second))

        if heap:
            return -heap[0]
        return 0

check(Solution().lastStoneWeight([2, 7, 4, 1, 8, 1]), 1)
check(Solution().lastStoneWeight([1]), 1)
check(Solution().lastStoneWeight([2, 2]), 0)
check(Solution().lastStoneWeight([10, 4]), 6)
```

### Task 4: how many are smaller?

```python
def count_smaller(nums, x):
    return bisect.bisect_left(nums, x)

check(count_smaller([1, 3, 3, 3, 7], 3), 1)
check(count_smaller([1, 3, 3, 3, 7], 4), 4)
check(count_smaller([1, 3, 3, 3, 7], 0), 0)
check(count_smaller([1, 3, 3, 3, 7], 10), 5)
```

### Task 5: how many times?

```python
def count_of(nums, x):
    return bisect.bisect_right(nums, x) - bisect.bisect_left(nums, x)

check(count_of([1, 3, 3, 3, 7], 3), 3)
check(count_of([1, 3, 3, 3, 7], 7), 1)
check(count_of([1, 3, 3, 3, 7], 5), 0)
check(count_of([], 5), 0)
```

### Task 6: LeetCode 35, Search Insert Position

The same loop as `my_bisect_left`, with one extra check: if the middle item is the target, return straight away.

```python
class Solution:
    def searchInsert(self, nums: list[int], target: int) -> int:
        left = 0
        right = len(nums) - 1
        while left <= right:
            mid = (left + right) // 2
            if nums[mid] == target:
                return mid
            if nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1
        return left

check(Solution().searchInsert([1, 3, 5, 6], 5), 2)
check(Solution().searchInsert([1, 3, 5, 6], 2), 1)
check(Solution().searchInsert([1, 3, 5, 6], 7), 4)
check(Solution().searchInsert([1, 3, 5, 6], 0), 0)
check(Solution().searchInsert([4], 4), 0)
```

With `bisect`, the whole method is one line: `return bisect.bisect_left(nums, target)`.

### Task 7: LeetCode 703, Kth Largest Element in a Stream

A heap of the k biggest, kept on `self`. Its smallest item, `self.heap[0]`, is the k-th largest.

```python
class KthLargest:
    def __init__(self, k: int, nums: list[int]):
        self.k = k
        self.heap = []
        for n in nums:
            self.add(n)

    def add(self, val: int) -> int:
        heapq.heappush(self.heap, val)
        if len(self.heap) > self.k:
            heapq.heappop(self.heap)
        return self.heap[0]

kl = KthLargest(3, [4, 5, 8, 2])
check(kl.add(3), 4)
check(kl.add(5), 5)
check(kl.add(10), 5)
check(kl.add(9), 8)
check(kl.add(4), 8)
```

`__init__` reuses `add` for the starting numbers, so the size-k rule lives in one place.
