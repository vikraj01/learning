## 1. Tidy a name

Split into words (which also throws away all the extra spaces), fix each word, then join with single spaces. An empty or all-spaces string gives no words, and joining an empty list gives `""`.

```python
def tidy_name(text):
    fixed = []
    for word in text.split():
        fixed.append(word[0].upper() + word[1:].lower())
    return " ".join(fixed)


check(tidy_name("  tANVI   kumar "), "Tanvi Kumar")
check(tidy_name("VIKASH"), "Vikash")
check(tidy_name("anna maria  de souza"), "Anna Maria De Souza")
check(tidy_name("o"), "O")
check(tidy_name("   "), "")
check(tidy_name(""), "")
```

## 2. Second best score

Two steps: find the best, then run "best so far" over the scores that are smaller than it. Starting the second box at `None` (not `0`) matters, because all the scores could be negative.

```python
def second_best(scores):
    if len(scores) == 0:
        return None
    best = max(scores)
    second = None
    for score in scores:
        if score < best:
            if second is None or score > second:
                second = score
    return second


check(second_best([5, 3, 5, 1]), 3)
check(second_best([1, 2]), 1)
check(second_best([-1, -5, -3]), -3)
check(second_best([4, 9, 9, 7, 9]), 7)
check(second_best([2, 2, 2]), None)
check(second_best([7]), None)
check(second_best([]), None)
```

## 3. Rotate the playlist

Shrink `k` with `%`, but only after checking for an empty list (`% 0` crashes). Then glue two slices: the tail and the head. When `k` is `0`, `songs[-0:]` is the **whole** list, so handle `k == 0` on its own.

```python
def rotate(songs, k):
    if len(songs) == 0:
        return []
    k = k % len(songs)
    if k == 0:
        return songs[:]
    return songs[-k:] + songs[:-k]


check(rotate([1, 2, 3, 4, 5], 2), [4, 5, 1, 2, 3])
check(rotate([1, 2, 3, 4, 5], 7), [4, 5, 1, 2, 3])
check(rotate([1, 2, 3, 4, 5], 5), [1, 2, 3, 4, 5])
check(rotate(["a", "b", "c"], 0), ["a", "b", "c"])
check(rotate([9], 4), [9])
check(rotate([], 3), [])
```

## 4. Bank statement

A running total with two "watchers" on it: one keeps the lowest value, one counts the bad moments. The order inside the loop matters: update the balance first, then check it.

```python
def statement(opening, transactions):
    balance = opening
    lowest = opening
    below = 0
    for t in transactions:
        balance = balance + int(t)
        if balance < lowest:
            lowest = balance
        if balance < 0:
            below = below + 1
    return [balance, lowest, below]


check(statement(0, ["+500", "-200", "-400", "+100"]), [0, -100, 1])
check(statement(100, ["-30", "+10"]), [80, 70, 0])
check(statement(50, []), [50, 50, 0])
check(statement(0, ["-10", "-10", "+5", "+20"]), [5, -20, 3])
check(statement(-5, ["+5"]), [0, -5, 0])
check(statement(10, ["+1000"]), [1010, 10, 0])
```

## 5. Squash the repeats

Keep the current run (its character and its length). When a different character arrives, write the run down and start a new one. The classic bug is forgetting the last run after the loop.

```python
def squash(text):
    if text == "":
        return ""
    result = ""
    current = text[0]
    count = 0
    for char in text:
        if char == current:
            count = count + 1
        else:
            result = result + current + str(count)
            current = char
            count = 1
    result = result + current + str(count)
    return result


check(squash("aaabcc"), "a3b1c2")
check(squash("aabaa"), "a2b1a2")
check(squash("AAa"), "A2a1")
check(squash("z"), "z1")
check(squash("  "), " 2")
check(squash("abc"), "a1b1c1")
check(squash(""), "")
```

## 6. Sentence report

Three steps: clean the words (and drop empty ones), find the longest with "best so far" (`>` keeps the first on a tie), then build the text, picking singular or plural.

```python
def report(sentence):
    words = []
    for raw in sentence.split():
        word = raw.strip(".,!?")
        if word != "":
            words.append(word)

    if len(words) == 0:
        return "0 words"

    longest = words[0]
    for word in words:
        if len(word) > len(longest):
            longest = word

    word_label = "words"
    if len(words) == 1:
        word_label = "word"
    letter_label = "letters"
    if len(longest) == 1:
        letter_label = "letter"
    return f"{len(words)} {word_label}, longest: {longest} ({len(longest)} {letter_label})"


check(report("I love warm tea in the morning."), "7 words, longest: morning (7 letters)")
check(report("Wait ... what?!"), "2 words, longest: Wait (4 letters)")
check(report("I"), "1 word, longest: I (1 letter)")
check(report("Hello,   world!"), "2 words, longest: Hello (5 letters)")
check(report("Is it raining? Yes, a lot."), "6 words, longest: raining (7 letters)")
check(report("   "), "0 words")
check(report("?! ..."), "0 words")
```

## 7. Password checker

One pass over the characters fills in the yes/no boxes, then a list is built rule by rule in the fixed order. The trick for capitals: only letters change under `lower()`/`upper()`, so `"4"` and `"!"` are neither.

```python
def password_problems(password):
    has_digit = False
    has_capital = False
    has_small = False
    has_space = False
    for char in password:
        if char.isdigit():
            has_digit = True
        if char != char.lower():
            has_capital = True
        if char != char.upper():
            has_small = True
        if char == " ":
            has_space = True

    problems = []
    if len(password) < 8:
        problems.append("too short")
    if not has_digit:
        problems.append("no digit")
    if not has_capital:
        problems.append("no capital")
    if not has_small:
        problems.append("no small letter")
    if has_space:
        problems.append("has a space")
    return problems


check(password_problems("Tea4Two!"), [])
check(password_problems("cats"), ["too short", "no digit", "no capital"])
check(password_problems("MY PASSWORD 1"), ["no small letter", "has a space"])
check(password_problems("12345678"), ["no capital", "no small letter"])
check(password_problems(""), ["too short", "no digit", "no capital", "no small letter"])
check(password_problems("Good pass 9"), ["has a space"])
```

## 8. Secret messages

Each letter is turned into a number (its position), moved, wrapped with `% 26`, and turned back into a letter. Working with the small version and fixing the case afterwards means you only need one alphabet.

```python
def caesar(text, shift):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    result = ""
    for char in text:
        position = alphabet.find(char.lower())
        if position == -1:
            result = result + char
        else:
            new_letter = alphabet[(position + shift) % 26]
            if char != char.lower():
                new_letter = new_letter.upper()
            result = result + new_letter
    return result


check(caesar("Hello, World!", 3), "Khoor, Zruog!")
check(caesar("xyz", 2), "zab")
check(caesar("Khoor", -3), "Hello")
check(caesar("abc", 29), "def")
check(caesar("Tea at 5pm?", 1), "Ufb bu 5qn?")
check(caesar("same", 0), "same")
check(caesar("", 4), "")
```

## 9. Numbers in words

Lists as lookup tables: the number itself is the index. `//` and `%` split a two-digit number into its tens and ones. Watch the spelling of `forty`.

```python
def to_words(n):
    small = ["zero", "one", "two", "three", "four", "five", "six", "seven", "eight", "nine",
             "ten", "eleven", "twelve", "thirteen", "fourteen", "fifteen", "sixteen",
             "seventeen", "eighteen", "nineteen"]
    tens = ["", "", "twenty", "thirty", "forty", "fifty", "sixty", "seventy", "eighty", "ninety"]

    if n < 20:
        return small[n]
    ten = n // 10
    one = n % 10
    if one == 0:
        return tens[ten]
    return tens[ten] + "-" + small[one]


check(to_words(0), "zero")
check(to_words(7), "seven")
check(to_words(13), "thirteen")
check(to_words(19), "nineteen")
check(to_words(20), "twenty")
check(to_words(42), "forty-two")
check(to_words(90), "ninety")
check(to_words(99), "ninety-nine")
```

## 10. Unsquash

Reading text into pieces is called **parsing**. Collect digits as text until a letter arrives, and only then turn them into a number with `int`. Same "don't forget the last one" trap as in problem 5.

```python
def unsquash(code):
    result = ""
    letter = ""
    digits = ""
    for char in code:
        if char.isdigit():
            digits = digits + char
        else:
            if letter != "":
                result = result + letter * int(digits)
            letter = char
            digits = ""
    if letter != "":
        result = result + letter * int(digits)
    return result


check(unsquash("a3b1c2"), "aaabcc")
check(unsquash("x12y1"), "xxxxxxxxxxxxy")
check(unsquash("z1"), "z")
check(unsquash("A2a1"), "AAa")
check(unsquash("b10"), "bbbbbbbbbb")
check(unsquash(""), "")
```

## 11. ⭐ Stretch: is this card number real?

Clean, validate, then compute. Reversing the text turns "every second digit from the right" into "every odd index", which is much easier to loop over.

```python
def valid_card(number):
    digits = number.replace(" ", "")
    if not digits.isdigit() or len(digits) < 2:
        return False

    total = 0
    for i, char in enumerate(digits[::-1]):
        d = int(char)
        if i % 2 == 1:
            d = d * 2
            if d > 9:
                d = d - 9
        total = total + d
    return total % 10 == 0


check(valid_card("4539 3195 0343 6467"), True)
check(valid_card("8273 1232 7352 0569"), False)
check(valid_card("059"), True)
check(valid_card("055 444 285"), True)
check(valid_card("055a 444 285"), False)
check(valid_card("0"), False)
check(valid_card(""), False)
```

## 12. ⭐ Stretch: wrap the text

A greedy loop: keep adding words while they fit, and close the line when one doesn't. The empty current line is the special case, both at the start of a line (no space) and at the end (don't add an empty line).

```python
def wrap(text, width):
    lines = []
    line = ""
    for word in text.split():
        if line == "":
            line = word
        elif len(line) + 1 + len(word) <= width:
            line = line + " " + word
        else:
            lines.append(line)
            line = word
    if line != "":
        lines.append(line)
    return lines


check(wrap("the quick brown fox jumps over the lazy dog", 10),
      ["the quick", "brown fox", "jumps over", "the lazy", "dog"])
check(wrap("tea with milk", 13), ["tea with milk"])
check(wrap("tea with milk", 12), ["tea with", "milk"])
check(wrap("a b c", 1), ["a", "b", "c"])
check(wrap("hello", 3), ["hello"])
check(wrap("hi supercalifragilistic yo", 5), ["hi", "supercalifragilistic", "yo"])
check(wrap("  spaced   out  ", 20), ["spaced out"])
check(wrap("", 5), [])
```
