### Task 1: count a letter

```python
class Solution:
    def countChar(self, s: str, ch: str) -> int:
        def count_from(i):
            if i == len(s):
                return 0
            rest = count_from(i + 1)
            if s[i] == ch:
                return 1 + rest
            return rest

        return count_from(0)

check(Solution().countChar("banana", "a"), 3)
check(Solution().countChar("tea", "z"), 0)
check(Solution().countChar("", "a"), 0)
```

### Task 2: find the big ones

```python
class Solution:
    def findBig(self, nums: list[int], limit: int) -> list[int]:
        found = []

        def visit(i):
            if i == len(nums):
                return
            if nums[i] > limit:
                found.append(nums[i])
            visit(i + 1)

        visit(0)
        return found

check(Solution().findBig([5, 12, 3, 20], 10), [12, 20])
check(Solution().findBig([1, 2], 10), [])
check(Solution().findBig([], 10), [])
```

### Task 3: add up the even numbers

```python
class Solution:
    def sumEvens(self, nums: list[int]) -> int:
        total = 0

        def visit(i):
            nonlocal total
            if i == len(nums):
                return
            if nums[i] % 2 == 0:
                total += nums[i]
            visit(i + 1)

        visit(0)
        return total

check(Solution().sumEvens([4, 7, 2, 8]), 14)
check(Solution().sumEvens([1, 3]), 0)
check(Solution().sumEvens([]), 0)
```

### Task 4: Fibonacci Number

Calling the method itself, with `self.`. (A helper with `@cache` works too, and is faster.)

```python
class Solution:
    def fib(self, n: int) -> int:
        if n < 2:
            return n
        return self.fib(n - 1) + self.fib(n - 2)

check(Solution().fib(0), 0)
check(Solution().fib(1), 1)
check(Solution().fib(2), 1)
check(Solution().fib(4), 3)
check(Solution().fib(10), 55)
```

### Task 5: Climbing Stairs

```python
from functools import cache

class Solution:
    def climbStairs(self, n: int) -> int:
        @cache
        def ways(k):
            if k <= 2:
                return k
            return ways(k - 1) + ways(k - 2)

        return ways(n)

check(Solution().climbStairs(1), 1)
check(Solution().climbStairs(2), 2)
check(Solution().climbStairs(3), 3)
check(Solution().climbStairs(5), 8)
check(Solution().climbStairs(45), 1836311903)
```

`if k <= 2: return k` covers both base cases: 1 way for 1 stair, 2 ways for 2 stairs.

### Task 6: N-th Tribonacci Number

```python
from functools import cache

class Solution:
    def tribonacci(self, n: int) -> int:
        @cache
        def t(k):
            if k == 0:
                return 0
            if k <= 2:
                return 1
            return t(k - 1) + t(k - 2) + t(k - 3)

        return t(n)

check(Solution().tribonacci(0), 0)
check(Solution().tribonacci(2), 1)
check(Solution().tribonacci(4), 4)
check(Solution().tribonacci(25), 1389537)
```

### Task 7: Add Digits

```python
class Solution:
    def addDigits(self, num: int) -> int:
        if num < 10:
            return num
        total = 0
        while num > 0:
            total += num % 10
            num = num // 10
        return self.addDigits(total)

check(Solution().addDigits(38), 2)
check(Solution().addDigits(0), 0)
check(Solution().addDigits(9), 9)
check(Solution().addDigits(199), 1)
```

### Quiz

1. **A**: `count += 1` means `count = count + 1`, a new value in the name, so Python needs `nonlocal` to know which `count`. `append` changes the list in place.
2. **B**: `@cache` saves each answer the first time and hands it back after that. It doesn't change the depth limit.
3. **C**: a method calls itself through `self`.
4. **A**: you're gathering many things along the way, so collect them into a list.
