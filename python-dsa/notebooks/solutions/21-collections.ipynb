### Task 1: count the letters

````python
def count_letters(word):
    counts = defaultdict(int)
    for ch in word:
        counts[ch] += 1
    return dict(counts)

check(count_letters("banana"), {"b": 1, "a": 3, "n": 2})
check(count_letters("tea"), {"t": 1, "e": 1, "a": 1})
check(count_letters(""), {})
````

`dict(Counter(word))` gives the same answer in one line.

### Task 2: group words by length

````python
def group_by_length(words):
    groups = defaultdict(list)
    for w in words:
        groups[len(w)].append(w)
    return dict(groups)

check(group_by_length(["cat", "dog", "tiger", "ox"]), {3: ["cat", "dog"], 5: ["tiger"], 2: ["ox"]})
check(group_by_length(["hi"]), {2: ["hi"]})
check(group_by_length([]), {})
````

### Task 3: the most common letter

`most_common(1)` is a list with one tuple, like `[('a', 3)]`. `[0]` takes the tuple, the next `[0]` takes the letter.

````python
def most_common_letter(word):
    return Counter(word).most_common(1)[0][0]

check(most_common_letter("banana"), "a")
check(most_common_letter("hello"), "l")
check(most_common_letter("zzz"), "z")
````

### Task 4: LeetCode 242, Valid Anagram

````python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        return Counter(s) == Counter(t)

check(Solution().isAnagram("anagram", "nagaram"), True)
check(Solution().isAnagram("rat", "car"), False)
check(Solution().isAnagram("a", "ab"), False)
````

### Task 5: LeetCode 383, Ransom Note

`need - have` is empty when you're short of nothing, and an empty Counter counts as `False`.

````python
class Solution:
    def canConstruct(self, ransomNote: str, magazine: str) -> bool:
        short = Counter(ransomNote) - Counter(magazine)
        return not short

check(Solution().canConstruct("a", "b"), False)
check(Solution().canConstruct("aa", "ab"), False)
check(Solution().canConstruct("aa", "aab"), True)
check(Solution().canConstruct("", "abc"), True)
````

### Task 6: LeetCode 2073, Time Needed to Buy Tickets

````python
class Solution:
    def timeRequiredToBuy(self, tickets: list[int], k: int) -> int:
        queue = deque()
        for person in range(len(tickets)):
            queue.append((person, tickets[person]))

        seconds = 0
        while queue:
            person, needed = queue.popleft()
            seconds += 1
            needed -= 1
            if needed == 0 and person == k:
                return seconds
            if needed > 0:
                queue.append((person, needed))

check(Solution().timeRequiredToBuy([2, 3, 2], 2), 6)
check(Solution().timeRequiredToBuy([5, 1, 1, 1], 0), 8)
check(Solution().timeRequiredToBuy([1], 0), 1)
````

There's also a shorter way without a queue, once you see the pattern: each person before or at `k` buys at most `tickets[k]` tickets, and each person after `k` buys at most `tickets[k] - 1` (person `k` finishes before their next turn).

```python
total = 0
for i in range(len(tickets)):
    if i <= k:
        total += min(tickets[i], tickets[k])
    else:
        total += min(tickets[i], tickets[k] - 1)
return total
```

### Task 7: LeetCode 933, Number of Recent Calls, the fast way

Old times only ever leave from the front, because every new `t` is bigger than the last. So the deque never holds more than the last 3000 milliseconds of calls.

````python
class RecentCounter:
    def __init__(self):
        self.times = deque()

    def ping(self, t: int) -> int:
        self.times.append(t)
        while self.times[0] < t - 3000:
            self.times.popleft()
        return len(self.times)

rc = RecentCounter()
check(rc.ping(1), 1)
check(rc.ping(100), 2)
check(rc.ping(3001), 3)
check(rc.ping(3002), 3)
check(rc.ping(7000), 1)
````
