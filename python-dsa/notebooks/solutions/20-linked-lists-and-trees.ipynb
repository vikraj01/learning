The classes, helpers and trees from the lesson:

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def build_list(values):
    head = None
    for v in values[::-1]:
        head = ListNode(v, head)
    return head

def to_list(head):
    values = []
    node = head
    while node:
        values.append(node.val)
        node = node.next
    return values

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

#        4
#      /   \
#     2     7
#    / \
#   1   3
small = TreeNode(4, TreeNode(2, TreeNode(1), TreeNode(3)), TreeNode(7))

single = TreeNode(5)          # one node, no children

#   1
#    \
#     2
#      \
#       3
chain = TreeNode(1, None, TreeNode(2, None, TreeNode(3)))

empty = None                  # an empty tree
````

### Task 2: how long is it?

````python
def length(head):
    count = 0
    node = head
    while node:
        count += 1
        node = node.next
    return count

check(length(build_list([4, 5, 6])), 3)
check(length(build_list([1])), 1)
check(length(None), 0)
````

### Task 3: is it in there?

Return `True` as soon as you find it; only after the whole walk do you know it isn't there.

````python
def contains(head, target):
    node = head
    while node:
        if node.val == target:
            return True
        node = node.next
    return False

check(contains(build_list([3, 1, 4]), 1), True)
check(contains(build_list([3, 1, 4]), 9), False)
check(contains(None, 5), False)
````

### Task 4: the last value

The empty list first, because `None.next` would crash. Then walk while there **is** a next node.

````python
def last_value(head):
    if not head:
        return None
    node = head
    while node.next:
        node = node.next
    return node.val

check(last_value(build_list([4, 5, 6])), 6)
check(last_value(build_list([9])), 9)
check(last_value(None), None)
````

### Task 5: LeetCode 876, Middle of the Linked List

Count, then walk `count // 2` steps. For 5 nodes that's 2 steps (to the 3); for 6 nodes it's 3 steps (to the 4, the second middle).

````python
class Solution:
    def middleNode(self, head: ListNode | None) -> ListNode | None:
        count = 0
        node = head
        while node:
            count += 1
            node = node.next

        node = head
        for step in range(count // 2):
            node = node.next
        return node

check(to_list(Solution().middleNode(build_list([1, 2, 3, 4, 5]))), [3, 4, 5])
check(to_list(Solution().middleNode(build_list([1, 2, 3, 4, 5, 6]))), [4, 5, 6])
check(to_list(Solution().middleNode(build_list([1]))), [1])
````

There's also a well-known trick with two walkers (notebook 13's two pointers): `slow` takes one step while `fast` takes two. When `fast` reaches the end, `slow` is in the middle.

```python
slow = fast = head
while fast and fast.next:
    slow = slow.next
    fast = fast.next.next
return slow
```

### Task 6: count the nodes in a tree

````python
def count_nodes(root):
    if not root:
        return 0
    return 1 + count_nodes(root.left) + count_nodes(root.right)

check(count_nodes(small), 5)
check(count_nodes(single), 1)
check(count_nodes(chain), 3)
check(count_nodes(empty), 0)
````

### Task 7: LeetCode 104, Maximum Depth of Binary Tree

````python
class Solution:
    def maxDepth(self, root: TreeNode | None) -> int:
        if not root:
            return 0
        return 1 + max(self.maxDepth(root.left), self.maxDepth(root.right))

check(Solution().maxDepth(small), 3)
check(Solution().maxDepth(single), 1)
check(Solution().maxDepth(chain), 3)
check(Solution().maxDepth(empty), 0)
````

Inside a class, the method calls itself as `self.maxDepth(...)`, because it lives on the object.

### Task 8: LeetCode 206, Reverse Linked List

````python
class Solution:
    def reverseList(self, head: ListNode | None) -> ListNode | None:
        prev = None
        node = head
        while node:
            nxt = node.next
            node.next = prev
            prev = node
            node = nxt
        return prev

check(to_list(Solution().reverseList(build_list([1, 2, 3, 4, 5]))), [5, 4, 3, 2, 1])
check(to_list(Solution().reverseList(build_list([1, 2]))), [2, 1])
check(to_list(Solution().reverseList(None)), [])
````
