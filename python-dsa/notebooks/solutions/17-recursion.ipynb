### Task 1: add up 1 to n

Base case: `n == 0` gives 0. Smaller step: `n + sum_to(n - 1)`.

```python
def sum_to(n):
    if n == 0:
        return 0
    return n + sum_to(n - 1)

check(sum_to(4), 10)
check(sum_to(1), 1)
check(sum_to(0), 0)
check(sum_to(10), 55)
```

### Task 2: power

Base case: `exp == 0` gives 1.

```python
def power(base, exp):
    if exp == 0:
        return 1
    return base * power(base, exp - 1)

check(power(2, 5), 32)
check(power(3, 0), 1)
check(power(5, 1), 5)
check(power(10, 3), 1000)
```

### Task 3: add up the digits, recursively

Base case: a single digit (less than 10) is its own digit sum. That also covers 0.

```python
def digit_sum(n):
    if n < 10:
        return n
    return n % 10 + digit_sum(n // 10)

check(digit_sum(4072), 13)
check(digit_sum(7), 7)
check(digit_sum(0), 0)
check(digit_sum(99), 18)
```

### Task 4: reverse a string

Reverse the rest, then put the first letter at the end.

```python
def reverse(s):
    if s == "":
        return ""
    return reverse(s[1:]) + s[0]

check(reverse("tea"), "aet")
check(reverse("a"), "a")
check(reverse(""), "")
check(reverse("Tanvi"), "ivnaT")
```

### Task 5: palindrome

```python
def is_palindrome(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return is_palindrome(s[1:-1])

check(is_palindrome("level"), True)
check(is_palindrome("abba"), True)
check(is_palindrome("tea"), False)
check(is_palindrome("a"), True)
check(is_palindrome(""), True)
```

### Task 6: the biggest number

```python
def biggest(nums):
    if len(nums) == 1:
        return nums[0]
    rest = biggest(nums[1:])
    if nums[0] > rest:
        return nums[0]
    return rest

check(biggest([3, 9, 2]), 9)
check(biggest([5]), 5)
check(biggest([-4, -1, -7]), -1)
check(biggest([1, 2, 3, 4]), 4)
```

### Task 7: Power of Two

Two base cases: 1 is yes; zero, negatives and odd numbers are no. Then halve.

```python
class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        if n == 1:
            return True
        if n <= 0 or n % 2 == 1:
            return False
        return self.isPowerOfTwo(n // 2)

check(Solution().isPowerOfTwo(1), True)
check(Solution().isPowerOfTwo(16), True)
check(Solution().isPowerOfTwo(3), False)
check(Solution().isPowerOfTwo(6), False)
check(Solution().isPowerOfTwo(0), False)
check(Solution().isPowerOfTwo(-8), False)
```
