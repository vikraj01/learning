### Task 1: first and last

`word[0]` is the first letter and `word[-1]` the last. For `"a"` both are the same letter.

```python
def first_and_last(word):
    return word[0] + word[-1]

check(first_and_last("python"), "pn")
check(first_and_last("Tanvi"), "Ti")
check(first_and_last("a"), "aa")
```

### Task 2: same both ways?

`word == word[::-1]` is already `True` or `False`, so return it directly.

```python
def is_palindrome(word):
    return word == word[::-1]

check(is_palindrome("level"), True)
check(is_palindrome("tea"), False)
check(is_palindrome("a"), True)
```

### Task 3: Defanging an IP Address

```python
class Solution:
    def defangIPaddr(self, address: str) -> str:
        return address.replace(".", "[.]")

check(Solution().defangIPaddr("1.1.1.1"), "1[.]1[.]1[.]1")
check(Solution().defangIPaddr("255.100.50.0"), "255[.]100[.]50[.]0")
```

### Task 4: Length of Last Word

`split()` throws away all the extra spaces, then `[-1]` is the last word.

```python
class Solution:
    def lengthOfLastWord(self, s: str) -> int:
        return len(s.split()[-1])

check(Solution().lengthOfLastWord("Hello World"), 5)
check(Solution().lengthOfLastWord("   fly me   to   the moon  "), 4)
check(Solution().lengthOfLastWord("a"), 1)
```

### Task 5: Find the Index of the First Occurrence

`find` gives the first index, or `-1` when it's not there: exactly what the problem asks.

```python
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
        return haystack.find(needle)

check(Solution().strStr("sadbutsad", "sad"), 0)
check(Solution().strStr("leetcode", "leeto"), -1)
check(Solution().strStr("hello", "ll"), 2)
```

### Task 6: Palindrome Number

Turn the number into text, then compare it with itself backwards. `"-121"` backwards is `"121-"`, so negatives are `False` without any extra code.

```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        text = str(x)
        return text == text[::-1]

check(Solution().isPalindrome(121), True)
check(Solution().isPalindrome(-121), False)
check(Solution().isPalindrome(10), False)
check(Solution().isPalindrome(7), True)
```
