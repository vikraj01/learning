# 9. Loop patterns

After this notebook you'll know the five loop shapes that solve most easy LeetCode problems: **total, counter, best so far, build a list, search**, plus **every pair** with a loop inside a loop.

Learn the shapes and you'll start seeing them in every problem. Each one has the same three parts:

1. **Set up** a box *before* the loop.
2. **Update** it *inside* the loop.
3. **Use** it *after* the loop.

When you get stuck on a problem, ask: "which shape is this? What's my box, and what does it start as?"

## Pattern 1: total

You did this in notebook 8. Box starts at `0`, add each item.

````python
marks = [70, 45, 90]
total = 0                 # set up
for m in marks:
    total += m            # update
print(total / len(marks)) # use: the average
````

```
68.33333333333333
```

`sum(marks)` gives the total too. The pattern still matters, because problems often want the total **at each step**, or a total of something you work out as you go.

## Pattern 2: counter

Box starts at `0`, add **1** each time something is true. The difference from a total: you add 1, not the item.

How many marks are a pass (40 or more)?

````python
marks = [70, 35, 90, 20, 55]
passed = 0
for m in marks:
    if m >= 40:
        passed += 1
    print("mark", m, "-> passed so far:", passed)
print(passed)
````

```
mark 70 -> passed so far: 1
mark 35 -> passed so far: 1
mark 90 -> passed so far: 2
mark 20 -> passed so far: 2
mark 55 -> passed so far: 3
3
```

| `m` | `m >= 40`? | `passed` after |
| --- | --- | --- |
| 70 | ✓ | 1 |
| 35 | ✗ | 1 |
| 90 | ✓ | 2 |
| 20 | ✗ | 2 |
| 55 | ✓ | 3 |

## Pattern 3: best so far

Keep the best thing you've seen so far, and replace it whenever you meet something better. Like walking past a row of mango stalls and remembering the cheapest price so far.

````python
nums = [3, 8, 5, 9, 2]
best = nums[0]
for n in nums:
    if n > best:
        best = n
    print("saw", n, "-> best so far:", best)
print(best)
````

```
saw 3 -> best so far: 3
saw 8 -> best so far: 8
saw 5 -> best so far: 8
saw 9 -> best so far: 9
saw 2 -> best so far: 9
9
```

`max(nums)` does this for you. You still need the pattern, because problems often want the best of something you **work out** as you go (the richest customer, the longest run), not the best item. For the smallest, flip `>` to `<`.

### The trap: what does `best` start as?

Starting at `0` looks natural. **Predict first:** what's the warmest temperature here?

````python
winter = [-5, -2, -8]
warmest = 0
for t in winter:
    if t > warmest:
        warmest = t
print(warmest)
````

```
0
```

Wrong: `0` isn't even in the list. Every temperature is below 0, so none of them ever beats the starting value. Start with **the first item** instead, `warmest = winter[0]`, so the start is a real value from the list. Fix the cell above and check you get `-2`.

## Pattern 4: build a new list

Start with an empty list `[]`, `append` to it. Often with an `if`, to keep only some items.

````python
prices = [120, 45, 300, 80]
cheap = []
for p in prices:
    if p < 100:
        cheap.append(p)
print(cheap)
````

```
[45, 80]
```

The same works for building a string: start with `""` and add to it with `+=`.

````python
word = "hello"
shout = ""
for ch in word:
    shout += ch.upper()
print(shout)
````

```
HELLO
```

## Pattern 5: search, and return as soon as you know

"Is there an even number?" You can stop the moment you find one. Inside a function, `return` ends the **whole function**, loop and all. So:

- `return True` **inside** the loop, when you find it.
- `return False` **after** the loop, when you've looked at everything and found nothing.

````python
def has_even(nums):
    for n in nums:
        if n % 2 == 0:
            return True
    return False

print(has_even([1, 3, 4, 7]))
print(has_even([1, 3, 5]))
````

```
True
False
```

### The classic bug: deciding too early

This one looks right, but there's an `else` inside the loop. **Predict first:** what does `has_even_wrong([1, 2])` give?

````python
def has_even_wrong(nums):
    for n in nums:
        if n % 2 == 0:
            return True
        else:
            return False

print(has_even_wrong([1, 2]))
````

```
False
```

It looked at `1`, saw it was odd, and returned `False` straight away. It never reached the `2`.

| `has_even_wrong([1, 2])` | |
| --- | --- |
| `n` is 1 | odd, so `else` → `return False` |
| `n` is 2 | never reached: the function already ended |

You can only say "yes" early. To say "no", you have to look at **everything** first, so that `return` goes **after** the loop.

## Two patterns together: a running total, saved at every step

Many problems mix two shapes. Here, a total (pattern 1) and building a list (pattern 4): your bank balance after each day.

````python
changes = [100, -30, 50, -20]
balance = 0
history = []
for c in changes:
    balance += c
    history.append(balance)
print(history)
````

```
[100, 70, 120, 100]
```

## Every pair: a loop inside a loop

Some problems need every **pair** of items: item 0 with item 1, item 0 with item 2, and so on. Put one loop inside another. For each `i`, the inner loop runs all the way through.

````python
nums = [5, 6, 7]
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        print(i, j, "->", nums[i], nums[j])
````

```
0 1 -> 5 6
0 2 -> 5 7
1 2 -> 6 7
```

| `i` | inner `range(i + 1, 3)` | pairs |
| --- | --- | --- |
| 0 | 1, 2 | (0, 1), (0, 2) |
| 1 | 2 | (1, 2) |
| 2 | nothing: `range(3, 3)` is empty | |

Why does `j` start at `i + 1`? Change the inner loop to `range(len(nums))` and run it again. You'll get 9 lines: each item paired with **itself** (`0 0`), and every pair **twice** (`0 1` and `1 0`). Starting at `i + 1` gives each pair once.

Every pair + a counter: how many pairs of marks add up to 100?

````python
marks = [40, 60, 50, 50, 30]
count = 0
for i in range(len(marks)):
    for j in range(i + 1, len(marks)):
        if marks[i] + marks[j] == 100:
            count += 1
print(count)
````

```
2
```

(40 + 60, and 50 + 50.)

## A shorter way: list comprehensions

Pattern 4 is so common that Python has a one-line way to write it, called a **list comprehension**:

````python
nums = [1, 2, 3, 4]
print([n * 2 for n in nums])
print([n for n in nums if n % 2 == 0])
````

```
[2, 4, 6, 8]
[2, 4]
```

Read the first one as "`n * 2`, for each `n` in `nums`". The second keeps only the items where the `if` is true. You'll see these in other people's solutions. Write the long way until it feels easy; then try the short way.
