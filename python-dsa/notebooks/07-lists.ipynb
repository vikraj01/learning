# 7. Lists

After this notebook you'll be able to keep many values in one place, in order, and change them.

Reading from a list works like strings, so that part is quick. Take your time on two traps: **`sort()` gives back nothing**, and **two names can share one list**.

## A shelf of numbered boxes

Picture a shelf with a row of boxes. Each box holds one thing, and each box has a number under it, starting at 0. You can look in any box, swap what's inside, add a box at the end, or take one away.

That shelf is a **list**. Most LeetCode problems hand you a list, usually called `nums`. When a problem says `nums: list[int]`, it means "a list of whole numbers".

```
 nums = [4, 8, 15, 16, 23]

  +----+----+----+----+----+
  |  4 |  8 | 15 | 16 | 23 |
  +----+----+----+----+----+
     0    1    2    3    4      <- index
    -5   -4   -3   -2   -1      <- negative index
```

Index, negative index, `len` and slicing work exactly like on strings:

````python
nums = [4, 8, 15, 16, 23]
print(nums[0])
print(nums[-1])
print(len(nums))
print(nums[1:3])
print(nums[::-1])
````

```
4
23
5
[8, 15]
[23, 16, 15, 8, 4]
```

A list can hold any kind of value: numbers, strings, `True`/`False`, even other lists.

## Changing a box

Here's the big difference from strings. A list **can** be changed in place:

````python
nums = [4, 8, 15, 16, 23]
nums[0] = 99
print(nums)
````

```
[99, 8, 15, 16, 23]
```

## Adding and taking away: `append`, `insert`, `pop`

Run this cell and follow the shelf as it changes:

````python
fruits = ["apple", "mango"]
fruits.append("kiwi")
print(fruits)
fruits.insert(0, "banana")
print(fruits)
last = fruits.pop()
print(last, fruits)
first = fruits.pop(0)
print(first, fruits)
print("mango" in fruits)
````

```
['apple', 'mango', 'kiwi']
['banana', 'apple', 'mango', 'kiwi']
kiwi ['banana', 'apple', 'mango']
banana ['apple', 'mango']
True
```

| Tool | What it does |
| --- | --- |
| `append(x)` | adds `x` at the end (you'll use this one most) |
| `insert(i, x)` | puts `x` at index `i`, pushes the rest along |
| `pop()` | takes the last item off and gives it to you |
| `pop(i)` | takes the item at index `i` |
| `x in nums` | `True` if `x` is somewhere in the list |

Two quick extras: `[1, 2] + [3, 4]` gives `[1, 2, 3, 4]`, and `[0] * 5` gives `[0, 0, 0, 0, 0]`. And `sum(nums)`, `max(nums)`, `min(nums)` give the total, the biggest and the smallest.

## Sorting: `sort()` and `sorted()`

There are two ways to sort, and the difference matters.

- **`nums.sort()`** sorts the list itself. The old order is gone.
- **`sorted(nums)`** leaves the list alone and gives you a **new**, sorted list.

````python
marks = [70, 45, 90, 60]
ordered = sorted(marks)
print(ordered)
print(marks)
marks.sort()
print(marks)
print(sorted(marks, reverse=True))
````

```
[45, 60, 70, 90]
[70, 45, 90, 60]
[45, 60, 70, 90]
[90, 70, 60, 45]
```

`reverse=True` means biggest first. It works with both.

### The trap: `sort()` gives back nothing

**Predict first:** what does this print?

````python
marks = [70, 45, 90, 60]
result = marks.sort()
print(result)
````

```
None
```

Remember `None` from notebook 4: "nothing came back". `sort()` changes the list and returns `None`, like a function with no `return`. So never write `marks = marks.sort()`. That sorts the list, then puts `None` in the box and loses the list. Run it to see what happens next:

````python
marks = [3, 1]
marks = marks.sort()
print(marks[0])
````

```
TypeError: 'NoneType' object is not subscriptable
```

`marks` is now `None`, and you can't use `[0]` on `None`. ("Subscriptable" means "you can use square brackets on it".) Write `marks.sort()` on its own line, or `marks = sorted(marks)`.

`nums.reverse()` is the same kind of tool: it flips the list itself and returns `None`.

## Lists inside lists

A list can hold other lists, like a shelf where each box holds a smaller shelf. Read `accounts[1][2]` left to right: first take box 1 (a whole list), then box 2 inside it.

````python
accounts = [[1, 2, 3], [3, 2, 1], [5, 5, 5]]
print(accounts[1])
print(accounts[1][2])
print(sum(accounts[2]))
````

```
[3, 2, 1]
1
15
```

## Two names, one list

This one surprises everyone. **Predict first:** what does the last line print?

````python
a = [1, 2, 3]
b = a
b.append(4)
print(a)
````

```
[1, 2, 3, 4]
```

You changed `b`, but `a` changed too. Why?

`b = a` does **not** make a second shelf. It sticks a second label on the **same** shelf:

```
 a = [1, 2, 3]          a ──┐
 b = a                      ├──> [1, 2, 3]       one shelf, two labels
                        b ──┘

 b.append(4)            a ──┐
                            ├──> [1, 2, 3, 4]    both labels see the change
                        b ──┘
```

If you want a real second shelf, copy it: `b = a.copy()` (or `b = a[:]`, a slice of everything).

````python
a = [1, 2, 3]
b = a.copy()
b.append(4)
print(a)
print(b)
````

```
[1, 2, 3]
[1, 2, 3, 4]
```

### The same thing happens with functions

When you pass a list to a function, the function gets the **same** shelf, not a copy. If it changes the boxes, the caller sees the change, even though the function returned nothing:

````python
def double_first(nums):
    nums[0] = nums[0] * 2

my_list = [5, 6, 7]
double_first(my_list)
print(my_list)
````

```
[10, 6, 7]
```

This matters on LeetCode. Some problems say **"Do not return anything, modify nums in-place"**. **In place** means "change the list you were given, don't make a new one". LeetCode then looks at your list, not at what you returned.

One catch. Changing the **boxes** changes the caller's list. Pointing the **name** at a new list does not. **Predict first:** what are the two prints?

````python
def reverse_wrong(nums):
    nums = nums[::-1]     # a NEW list, only the name inside this function moves

def reverse_right(nums):
    nums.reverse()        # changes the shelf itself

a = [1, 2, 3]
reverse_wrong(a)
print(a)
reverse_right(a)
print(a)
````

```
[1, 2, 3]
[3, 2, 1]
```

```
 reverse_wrong:   nums = nums[::-1]

   a ────> [1, 2, 3]       (untouched)
   nums ─> [3, 2, 1]       a new shelf; thrown away when the function ends
```

For in-place problems, use things that change the list itself: `nums[i] = ...`, `append`, `pop`, `sort()`, `reverse()`.

## Common mistakes

**An index past the end.** `[4, 8, 15]` has indexes 0, 1, 2. This is the error you'll see most often with lists:

````python
nums = [4, 8, 15]
print(nums[3])
````

```
IndexError: list index out of range
```

"There's no box with that number." The last index is `len(nums) - 1`.
