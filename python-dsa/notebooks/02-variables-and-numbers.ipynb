# 2. Variables and numbers

After this notebook you'll be able to store values under a name and do maths with them, including the two operators LeetCode loves: `//` and `%`.

## A labelled box

Think of a box with a label on it. The label says `age`, and inside is `20`. Later you ask "what's in the `age` box?" and get `20` back.

In Python that box is a **variable**. You make one with `=`. Run this:

````python
age = 20
print(age)
````

```
20
```

- `age` is the **name** (the label), `20` is the **value** (what's inside).
- `=` means "put this value in this box". Read `age = 20` as "age gets 20". It's not "equals" from maths.
- Names use letters, numbers and `_`, with no spaces, and can't start with a number. `my_age` is fine, `my age` and `2age` are not.

**Boxes stay between cells.** The notebook remembers `age` now, so another cell can use it:

````python
print(age + 1)
````

```
21
```

That's why you run cells from top to bottom. If you see `NameError: name 'age' is not defined`, you skipped the cell that made the box. Run it first.

## Kinds of values

Every value has a kind, called its **type**. Four you'll use all the time:

- **int** (integer): a whole number, like `20`, `0`, `-7`.
- **float**: a number with a decimal point, like `49.5` or `3.0`.
- **str** (string): text, always in quotes, like `"Tanvi"`.
- **bool** (boolean): `True` or `False`, with capital letters. More on these in the next notebook.

`type()` tells you which kind a value is:

````python
price = 49.5
name = "Tanvi"
is_raining = False

print(type(age))
print(type(price))
print(type(name))
print(type(is_raining))
````

```
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

## Maths

Run this and match each answer to its line:

````python
print(7 + 2)    # add
print(7 - 2)    # subtract
print(7 * 2)    # multiply
print(7 / 2)    # divide
print(7 // 2)   # divide, keep only the whole part
print(7 % 2)    # remainder: what's left over
print(7 ** 2)   # power: 7 times 7
````

```
9
5
14
3.5
3
1
49
```

Brackets work like in school: `(2 + 3) * 4` is `20`, and `2 + 3 * 4` is `14`.

One thing to remember: **`/` always gives a float**. `10 / 2` is `5.0`, not `5`. If a LeetCode problem wants a whole number back, use `//`.

## `//` and `%`: sharing sweets

This is the part worth slowing down for. These two show up in LeetCode problems all the time.

You have 7 sweets and 2 friends. Each friend gets 3, and 1 is left over.

```
  friend 1:  ● ● ●
  friend 2:  ● ● ●
  left over: ●

  7 // 2  ->  3   (how many each)
  7 %  2  ->  1   (how many left)
```

- **`//`** is **floor division**: how many whole times it fits.
- **`%`** is **modulo** (say "mod"): what's left over.

**Predict, then run.** Before running the cell below, guess the four answers. Write them on paper if you like.

````python
print(17 // 5)
print(17 % 5)
print(20 // 4)
print(20 % 4)
````

```
3
2
5
0
```

17 sweets for 5 friends: 3 each, 2 left. 20 for 4: 5 each, nothing left. A remainder of `0` means "it divides exactly".

### Three uses you'll meet again

**Minutes into hours and minutes.** A film is 135 minutes long:

````python
minutes = 135
print(minutes // 60, "hours and", minutes % 60, "minutes")
````

```
2 hours and 15 minutes
```

**Even or odd.** An even number splits into 2 with nothing left over, so `% 2` is `0` for even and `1` for odd:

````python
print(14 % 2)
print(7 % 2)
````

```
0
1
```

**Taking a number apart.** `% 10` gives the last digit, and `// 10` drops it:

````python
number = 1234
print(number % 10)
print(number // 10)
````

```
4
123
```

```
   1 2 3 | 4
     │     └── 1234 %  10  ->  4     (the last digit)
     └──────── 1234 // 10  ->  123   (everything except the last digit)
```

**Try it:** change `number` to `9876` and `minutes` to `200`. Guess the answers before you run them.

## Updating a box

Python works out the right side first, then puts the answer in the box:

````python
score = 10
score = score + 1
print(score)

score += 5      # short for: score = score + 5
print(score)
````

```
11
16
```

`score = score + 1` reads "take what's in score, add 1, put it back". `-=`, `*=` and `//=` work the same way as `+=`.

## Rounding

`round(value, places)` rounds a float to that many places after the point:

````python
print(7 / 3)
print(round(7 / 3, 2))
print(round(7 / 3))
````

```
2.3333333333333335
2.33
2
```

Without the second number, `round` gives a whole number.

## Common mistakes

**Adding text and a number.** Run it and read the last line:

````python
print("5" + 3)
````

```
TypeError: can only concatenate str (not "int") to str
```

A **TypeError** means the kinds don't fit together. `"5"` has quotes, so it's text, and Python won't add text to a number. Remove the quotes and `5 + 3` is `8`.

**Dividing by zero** gives `ZeroDivisionError: division by zero`, with `/`, `//` and `%` alike. Same as in maths.
