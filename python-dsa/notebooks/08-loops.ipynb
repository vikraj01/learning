## A playlist, one song at a time

A music app plays a playlist like this: take the first song, play it. Take the next song, play it. Keep going until there are no songs left.

A **loop** is the same idea in code: "for each thing in this group, do this". You write the work once, and Python repeats it for every item.

## `for`: do something for each item

````python
songs = ["Kesariya", "Tum Hi Ho", "Ilahi"]
for song in songs:
    print("Now playing:", song)
print("Playlist over")
````

```
Now playing: Kesariya
Now playing: Tum Hi Ho
Now playing: Ilahi
Playlist over
```

How to read it:

1. `for song in songs:` means "take each item of `songs`, one at a time, and call it `song`".
2. The indented lines under it are the **body** of the loop. They run once per item, like the body of an `if`.
3. When there are no items left, the loop ends and Python carries on with the first line that isn't indented.

Each time round the loop is called an **iteration**. Here is what happens in each one:

| Iteration | `song` is | Printed |
| --- | --- | --- |
| 1 | `"Kesariya"` | `Now playing: Kesariya` |
| 2 | `"Tum Hi Ho"` | `Now playing: Tum Hi Ho` |
| 3 | `"Ilahi"` | `Now playing: Ilahi` |
| after the loop | | `Playlist over` |

You can pick any name instead of `song`. Pick one that says what the item is.

## A box that changes each time round

Loops get useful when they **remember** something from one iteration to the next. Make a box **before** the loop, change it **inside** the loop, use it **after** the loop.

Add up the prices in a shopping basket. The `print` inside is only there so you can watch:

````python
prices = [40, 25, 10]
total = 0
for p in prices:
    total += p
    print("added", p, "-> total is", total)
print("Pay:", total)
````

```
added 40 -> total is 40
added 25 -> total is 65
added 10 -> total is 75
Pay: 75
```

(`total += p` is short for `total = total + p`, from notebook 2.)

| Iteration | `p` | `total` before | `total` after |
| --- | --- | --- | --- |
| 1 | 40 | 0 | 40 |
| 2 | 25 | 40 | 65 |
| 3 | 10 | 65 | 75 |

Why `total = 0` **before** the loop? Try moving it inside the loop (indent it, above `total += p`) and run again. Each iteration would empty the box, and you'd only get the last price. The next notebook is all about these "remember something" patterns.

## `range`: counting numbers

Sometimes you don't have a list; you want the numbers 0, 1, 2, ... **`range`** makes them. (`list(...)` turns it into a list only so you can see the numbers. In a loop you don't need it.)

**Predict first:** write down the four lists, then run.

````python
print(list(range(5)))
print(list(range(2, 6)))
print(list(range(0, 10, 2)))
print(list(range(4, -1, -1)))
````

```
[0, 1, 2, 3, 4]
[2, 3, 4, 5]
[0, 2, 4, 6, 8]
[4, 3, 2, 1, 0]
```

The rule is the same as slicing: **it stops before the end number**.

| You write | You get | The end number |
| --- | --- | --- |
| `range(5)` | 0, 1, 2, 3, 4 (five numbers) | 5 is not included |
| `range(2, 6)` | 2, 3, 4, 5 | 6 is not included |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 (jump by 2) | 10 is not included |
| `range(4, -1, -1)` | 4, 3, 2, 1, 0 (counting down) | -1 is not included |

So to count **1 to 10**, you write `range(1, 11)`. That `+ 1` is easy to forget.

**Try it:** in the cell below, make it print 1 to 10.

````python
for i in range(1, 10):   # fix me: should print 1 to 10
    print(i)
````

## Index and item: `range(len(nums))` and `enumerate`

`for n in nums` gives you each **item**, but not its **position**. Sometimes you need the position too: to change `nums[i]`, to look at the next item `nums[i + 1]`, or to say *where* something was.

`range(len(nums))` gives you every index of the list. `len(nums)` is 3, so `range(3)` gives 0, 1, 2: exactly the indexes.

````python
nums = [10, 20, 30]
for i in range(len(nums)):
    print("index", i, "holds", nums[i])
````

```
index 0 holds 10
index 1 holds 20
index 2 holds 30
```

**`enumerate`** does the same job more neatly. It hands you two things each time round, the index and the item, and you name both:

````python
nums = [10, 20, 30]
for i, n in enumerate(nums):
    print("index", i, "holds", n)
````

```
index 0 holds 10
index 1 holds 20
index 2 holds 30
```

| Loop | Each time round you get |
| --- | --- |
| `for n in nums:` | the item |
| `for i in range(len(nums)):` | the index (get the item with `nums[i]`) |
| `for i, n in enumerate(nums):` | both |

Use whichever reads more clearly to you.

## `while`: repeat as long as something is true

A `for` loop goes through a group of things. A **`while`** loop keeps going **as long as** a condition is `True`. Like stirring tea *while* the sugar hasn't dissolved: you don't know in advance how many stirs.

Halve a number until it reaches zero, and count the steps:

````python
n = 20
steps = 0
while n > 0:
    n = n // 2
    steps += 1
    print("n is now", n)
print("steps:", steps)
````

```
n is now 10
n is now 5
n is now 2
n is now 1
n is now 0
steps: 5
```

Python checks `n > 0` **before** each iteration:

| Check `n > 0` | `n` after `n // 2` | `steps` |
| --- | --- | --- |
| 20 > 0 ✓ | 10 | 1 |
| 10 > 0 ✓ | 5 | 2 |
| 5 > 0 ✓ | 2 | 3 |
| 2 > 0 ✓ | 1 | 4 |
| 1 > 0 ✓ | 0 | 5 |
| 0 > 0 ✗ | loop stops | |

### Taking a number apart, digit by digit

From notebook 2: `% 10` gives the last digit, and `// 10` removes it. Put those in a `while` loop and you get every digit. You'll use this whenever a problem talks about "the digits of a number".

````python
n = 472
while n > 0:
    digit = n % 10
    print("digit:", digit)
    n = n // 10
````

```
digit: 2
digit: 7
digit: 4
```

| `n` | `n % 10` (digit) | `n // 10` (new `n`) |
| --- | --- | --- |
| 472 | 2 | 47 |
| 47 | 7 | 4 |
| 4 | 4 | 0 → loop stops |

The digits come out from the back.

### A loop that never ends

If the condition never becomes `False`, a `while` loop runs forever. The cell below forgot to change `n`. Run it on purpose:

1. The brackets on the left show `[*]`. The star means "still running".
2. Stop it: click the **■** (square) button in the toolbar at the top, or use the menu **Kernel → Interrupt Kernel**.
3. You'll see a red `KeyboardInterrupt` error. That only means "you stopped me". Nothing is broken.

````python
# This loop never ends: nothing inside changes n.
n = 3
steps = 0
while n > 0:
    steps += 1
````

Whenever a loop seems stuck, interrupt it, then check that something inside the loop moves you towards the end (here, `n = n - 1` is missing).

**Which loop when?** Use `for` when you have a group of things (a list, a string, a `range`). Use `while` when you keep going *until something happens* and don't know how many times.

## `break` and `continue`

- **`break`** stops the loop straight away.
- **`continue`** skips the rest of this iteration and moves on to the next item.

Look for the first negative number. **Predict first:** is `5` ever checked?

````python
nums = [4, 7, 9, -2, 5]
for n in nums:
    if n < 0:
        print("Found a negative:", n)
        break
    print("checked", n)
````

```
checked 4
checked 7
checked 9
Found a negative: -2
```

The `5` is never checked. Once you've found what you were looking for, there's no need to keep going.

````python
for n in [1, 2, 3, 4, 5]:
    if n % 2 == 0:
        continue
    print(n)
````

```
1
3
5
```

For even numbers, `continue` jumps back to the top before the `print`.

## Common mistakes

**Forgetting to indent the body.** Run it and read the last line:

````python
for n in [1, 2]:
print(n)
````

```
IndentationError: expected an indented block after 'for' statement on line 1
```

The lines that belong to the loop must be indented (4 spaces), like under an `if`.

**Using the item as if it were an index.**

````python
nums = [10, 20, 30]
for n in nums:
    print(nums[n])
````

```
IndexError: list index out of range
```

`n` is the item (10, 20, 30), not the position, and `nums[10]` doesn't exist. Either print `n`, or loop with `range(len(nums))` if you need positions.

**Looking one step too far.** This compares each item with the next one:

````python
nums = [3, 5, 8]
for i in range(len(nums)):
    print(nums[i], "then", nums[i + 1])
````

```
IndexError: list index out of range
```

The first two lines print, then it crashes. On the last iteration `i` is 2, so `nums[i + 1]` is `nums[3]`, past the end. If you need `nums[i + 1]`, loop with `range(len(nums) - 1)`. Fix the cell above and run it again.
