## A recipe card

Think of a recipe card for chai. It has a name ("Chai"), what you need to bring (milk, sugar), and the steps. Once it's written, anyone can say "make chai" with their own milk and sugar, and get a cup back.

A **function** is a recipe card in Python. You write the steps once, give them a name, and use them as often as you like with different inputs.

## Writing one: `def`

````python
def add(a, b):
    return a + b
````

Running that cell shows nothing. Writing a recipe card doesn't make any chai. But Python now knows a function called `add`.

```
def add(a, b):        <- def = "define a function". Its name is add.
    return a + b      <- the step: work out a + b and hand it back
```

- **`def`** means "I'm writing a new function".
- **`add`** is its name. Same naming rules as variables.
- **`a` and `b`** are its **parameters**: empty boxes for the inputs, filled in each time the function is used.
- The **colon** and the **indented** lines work like `if`. The indented lines are the function's steps.
- **`return`** hands an answer back to whoever asked.

## Using one: calling

Using a function is called **calling** it: its name, with the inputs in brackets.

````python
total = add(2, 3)
print(total)
print(add(10, 20))
````

```
5
30
```

What happens at `add(2, 3)`, step by step:

| Step | What Python does |
| --- | --- |
| 1 | jumps into `add` and fills the boxes: `a = 2`, `b = 3` |
| 2 | works out `a + b`, which is `5` |
| 3 | `return` hands `5` back and leaves the function |
| 4 | back outside, `add(2, 3)` is replaced by `5`, so `total = 5` |

The values you pass in (`2` and `3`) are called **arguments**. They fill the parameters in order: the first goes into `a`, the second into `b`.

You've been calling functions since notebook 1: `print(...)`, `type(...)` and `round(...)` are functions that come with Python.

**Try it:** add a cell with `print(add(100, -1))`. What do you expect?

## `print` and `return` are different

This is the most important part of the notebook. It trips up almost every beginner, and on LeetCode it's the difference between a right and a wrong answer.

- **`print`** shows something on the screen, for a person to read. The code can't use it afterwards.
- **`return`** hands a value back to the code that called the function. The code can store it, do maths with it, or check it.

You ask a friend to add up a bill. If they **shout** the total across the room (`print`), you heard it, but you have nothing in your hand. If they **write it on a slip and give it to you** (`return`), you can keep it, add a tip, or pass it on.

Here are two functions that look almost the same:

````python
def add_and_print(a, b):
    print(a + b)

def add_and_return(a, b):
    return a + b
````

**Predict, then run.** The next cell calls each one and stores the result. Before you run it, guess: what will be in `x`, and what will be in `y`?

````python
x = add_and_print(2, 3)
y = add_and_return(2, 3)
print("x is", x)
print("y is", y)
````

```
5
x is None
y is 5
```

Line by line:

| Line | What's on screen | What goes in the box |
| --- | --- | --- |
| `x = add_and_print(2, 3)` | `5` (the function printed it) | nothing came back, so `x` is `None` |
| `y = add_and_return(2, 3)` | nothing | `5` came back, so `y` is `5` |
| `print("x is", x)` | `x is None` | |
| `print("y is", y)` | `y is 5` | |

**`None`** is Python's word for "nothing". A function with no `return` gives back `None`.

The `5` on the screen *looked* like an answer from `add_and_print`. It wasn't. It was shouted, and the code never got it.

### A notebook trap: `Out[ ]`

A notebook shows the value of the **last line** of a cell, with `Out[...]` next to it. Run these two cells:

````python
add_and_return(4, 4)
````

````python
add_and_print(4, 4)
````

The first shows `Out[...]: 8`. That's the notebook being helpful: it displays what came back. The second shows `8` with no `Out`, because it was printed, and nothing came back.

Outside a notebook (and on LeetCode), there's no `Out`. Only `print` shows things on screen. So don't let `Out` fool you into thinking a value was printed.

### Why this matters for LeetCode

LeetCode doesn't read your screen. It calls your function and checks what it **returns**. If you print the right answer and don't return it, LeetCode gets `None` and marks you wrong.

- Use **`return`** for the answer.
- Use `print` only to look at things while you work something out.

## `return` ends the function

As soon as Python reaches a `return`, it hands the value back and leaves. Lines after it don't run:

````python
def say_hi():
    return "hi"
    print("you'll never see this")

print(say_hi())
````

```
hi
```

`say_hi` has no parameters, so the brackets are empty, both in `def say_hi():` and in `say_hi()`.

## Returning True or False

This ending-early is useful with `if`. Here's the even/odd check as a function:

````python
def is_even(n):
    if n % 2 == 0:
        return True
    return False

print(is_even(4))
print(is_even(7))
````

```
True
False
```

| Call | `n % 2 == 0`? | What happens |
| --- | --- | --- |
| `is_even(4)` | `True` | `return True`, leave. The last line never runs. |
| `is_even(7)` | `False` | skip the `if`, reach `return False` |

There's no `else`, because `return True` has already left the function. Many LeetCode problems say "return true if ...", and they look exactly like this.

## One function calling another

````python
def is_odd(n):
    return not is_even(n)

print(is_odd(7))
````

```
True
```

`is_odd(7)` calls `is_even(7)`, which returns `False`, and `not False` is `True`.

## Boxes inside a function stay inside

A variable made inside a function only exists while the function runs. Run this, and read the last line:

````python
def price_with_tip(bill):
    tip = bill // 10
    return bill + tip

print(price_with_tip(200))
print(tip)
````

```
220
...
NameError: name 'tip' is not defined
```

`tip` was a scratch box inside the function, thrown away when it finished. Only the returned value (`220`) comes out. If you need something outside, return it.

## Common mistakes

**Printing instead of returning.** No error message, which makes it sneaky. If you see `None` where you expected a value, look for a missing `return`.

**The wrong number of inputs.** Run it:

````python
print(add(2))
````

```
TypeError: add() missing 1 required positional argument: 'b'
```

`add` has two parameters but got one value. Pass one value per parameter.

**Calling a function before its `def` cell has run.** You get a `NameError`. Run the `def` cell first.
