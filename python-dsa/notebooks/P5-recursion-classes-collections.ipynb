````python
# Run this cell before problems 10 and 12.
# ListNode and TreeNode are the same as in notebook 20.

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_list(values):
    # ["a", "b"] -> a linked list a -> b
    head = None
    for v in values[::-1]:
        head = ListNode(v, head)
    return head

def to_list(head):
    # a linked list -> a normal Python list, so the checks are easy to read
    values = []
    node = head
    while node:
        values.append(node.val)
        node = node.next
    return values
````

Then `insort` puts the new one in its place. `top` can sort the dictionary's names with a `key`, like problem 2.
