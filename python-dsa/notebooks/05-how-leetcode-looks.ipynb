## A workbook with an answer checker

Remember maths workbooks with the answers at the back? **LeetCode** (leetcode.com) is a website full of programming questions, called **problems**, with the checker built in. You write a function, press a button, and LeetCode runs it on lots of inputs and tells you if every answer was right.

## What a problem page looks like

Here's [2235. Add Two Integers](https://leetcode.com/problems/add-two-integers/):

```
┌─────────────────────────────────┬─────────────────────────────────────┐
│ 2235. Add Two Integers   Easy   │  Python3 ▾                          │
│                                 │                                     │
│ Given two integers num1 and     │  class Solution:                    │
│ num2, return the sum of the     │      def sum(self, num1: int,       │
│ two integers.                   │              num2: int) -> int:     │
│                                 │                                     │
│ Example 1:                      │                                     │
│   Input: num1 = 12, num2 = 5    │                                     │
│   Output: 17                    │                                     │
│                                 ├─────────────────────────────────────┤
│ Constraints:                    │  Testcase  |  Test Result           │
│   -100 <= num1, num2 <= 100     │             [ Run ]   [ Submit ]    │
└─────────────────────────────────┴─────────────────────────────────────┘
```

On the left:

- **Description**: what your function gets and what it must return. Read it twice.
- **Examples**: inputs and the right output. Work one through on paper before writing code.
- **Constraints**: limits on the inputs. Here both numbers are between -100 and 100.

On the right:

- A language menu. Pick **Python3**.
- The start of a function, already written. This is the **template**. You fill in the inside.
- **Run** tries your code on the examples. **Submit** runs it on all the hidden tests (often hundreds).

Two words you'll see: an **integer** is a whole number (an `int`), and an **array** is a list of values (lists are notebook 7). LeetCode writes `true` and `false` in small letters; in Python they're always `True` and `False`.

## `class Solution`: the box your function lives in

```python
class Solution:
    def sum(self, num1: int, num2: int) -> int:
```

You know `def`. Two new things:

- **`class Solution:`** is a box LeetCode puts your function in. Every problem has it. You don't need to understand classes yet (that's notebook 19). Leave the line as it is and write your function indented inside it.
- **`self`** is always the first parameter. Python fills it in by itself. You never pass anything for it, and you won't use it in these problems.

So, with `self` set aside, this is a function `sum` with two inputs, `num1` and `num2`. Exactly like `add(a, b)` from notebook 4.

## Type hints: labels on the boxes

```
def sum(self, num1: int, num2: int) -> int:
                    ───        ───     ───
                     │          │       └── it returns an int
                     │          └────────── num2 is an int
                     └───────────────────── num1 is an int
```

The `: int` and `-> int` parts are **type hints**: labels saying what goes in and what comes out. Python doesn't check them; they're there for you. Read them first, every time:

- `-> int`: return a whole number
- `-> float`: return a number with a decimal point
- `-> bool`: return `True` or `False`
- `-> str`: return text
- `list[int]`: a list of whole numbers (notebook 7)

## Calling a function that lives in the box

To call it yourself, first make the box with `Solution()`, then call the function with a dot:

````python
class Solution:
    def double(self, n: int) -> int:
        return n * 2

print(Solution().double(4))
print(Solution().double(-3))
````

```
8
-6
```

- `Solution()` makes the box.
- `.double(4)` calls `double` inside it, with `n = 4`. You don't pass `self`.

Look at the indentation: `def` is 4 spaces in (inside `class`), and `return` is 8 spaces in (inside `def`).

### One ✓ is not enough

Suppose you'd written `return num1 - num2` by mistake. Then:

```
✗ Expected 17, but got 7
✗ Expected -6, but got -14
✓ Correct: 0
```

`(0, 0)` passed by luck, because `0 - 0` is also `0`. A solution is only right when **every** test is ✓. When one fails, take that input and work it through by hand.

## Submit on LeetCode

Once every line is ✓:

1. Open the problem on LeetCode and sign in.
2. Check the language says **Python3**.
3. Copy only the `class Solution` part: from `class Solution:` to your `return` line. Not the check lines.
4. Paste it into LeetCode's editor, replacing what's there.
5. Press **Run**, then **Submit**.

You'll see one of these:

- **Accepted**: every hidden test passed.
- **Wrong Answer**: LeetCode shows the input that failed, your output, and the expected one. Work that input through by hand and fix it.
- **Runtime Error**: your code crashed. Read the last line of the error, as in notebook 1.

Wrong answers are normal, and nobody sees them but you. Submitting several times is fine.

## Common mistakes

**`def` not indented under `class`.** You get `IndentationError: expected an indented block after class definition`. `def` goes 4 spaces in, its steps 8.

**Every test says `got None`.** You left `pass`, or used `print` instead of `return`.

**Renaming the function.** LeetCode calls it by its exact name (`sum` here). If you rename it to `add`, LeetCode can't find it. Keep the name and the parameters exactly as given.
