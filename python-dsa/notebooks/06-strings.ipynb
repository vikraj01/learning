# 6. Strings

After this notebook you'll be able to pick letters out of text, cut pieces from it, and change it with Python's ready-made tools.

Most of this is quick. Two parts need care: **slicing** (where a piece stops) and **strings can't be changed**. Slow down there.

## A necklace of letter beads

Think of a word as a necklace of beads, one letter on each bead. The beads are in a fixed order, and you can count them.

In Python, text is called a **string** (a string of letters, like a string of beads). You've met it already: text inside quotes.

## Index: one letter at a time

Every letter has a position number, called its **index**. Python starts counting at **0**, not 1.

```
 word = "python"

   p    y    t    h    o    n
   0    1    2    3    4    5      <- index from the front
  -6   -5   -4   -3   -2   -1      <- index from the back
```

Put the index in square brackets to get that letter. Run this:

````python
word = "python"
print(word[0])
print(word[5])
print(word[-1])
print(word[-2])
print(len(word))
````

```
p
n
n
o
6
```

- `word[0]` is the first letter.
- A **negative index** counts from the back: `word[-1]` is always the last letter, `word[-2]` the one before it.
- **`len()`** (short for length) tells you how many letters there are. Spaces count too.

The last index is `5`, but the length is `6`. The last index is always `len(word) - 1`, because counting starts at 0.

## Slicing: cut out a piece

A **slice** is a piece of the string. You write `[start:stop]`. You get the letters from `start` up to, **but not including**, `stop`.

This "not including" is the part everyone gets wrong at first. Picture the index numbers sitting on the beads, and the slice stopping **before** the `stop` bead:

```
 word[1:4]

   p    y    t    h    o    n
   0   [1    2    3]   4    5
        ^              ^
      start          stop (not taken)

 -> "yth"   (3 letters: 4 - 1 = 3)
```

A quick check: the slice has `stop - start` letters.

**Predict first:** what will each line print? Write your guesses down, then run it.

````python
word = "python"
print(word[1:4])
print(word[0:3])
print(word[2:])
print(word[:4])
print(word[::-1])
````

```
yth
pyt
thon
pyth
nohtyp
```

| Slice | Indexes taken | Result |
| --- | --- | --- |
| `word[1:4]` | 1, 2, 3 | `yth` |
| `word[0:3]` | 0, 1, 2 | `pyt` |
| `word[2:]` | 2 to the end | `thon` |
| `word[:4]` | the start to 3 | `pyth` |
| `word[::-1]` | all, backwards | `nohtyp` |

- Leave out the start and Python starts at the beginning. Leave out the stop and it goes to the end.
- `word[::-1]` is the string **backwards**. The third number is the **step**: how far to move each time. `-1` means "walk backwards". You'll use this one a lot.

**Try it:** in the cell below, get `"tho"` out of `"python"`. Then get `"on"` using a negative index.

````python
word = "python"
print(word[0:1])   # change this
````

## Methods: tools that come with every string

A **method** is a tool that belongs to a value. You use it with a dot: `value.method()`.

````python
name = "  Tanvi Kumar  "
print(name.lower())
print(name.upper())
print(name.strip())
print("1.1.1.1".replace(".", "[.]"))
print("hello".find("l"))
print("hello".find("z"))
print("th" in "python")
````

```
  tanvi kumar
  TANVI KUMAR
Tanvi Kumar
1[.]1[.]1[.]1
2
-1
True
```

| Tool | What it does |
| --- | --- |
| `lower()` / `upper()` | every letter small / capital |
| `strip()` | removes spaces from both ends (not the middle) |
| `replace(old, new)` | swaps every `old` piece for `new` |
| `find(piece)` | index where `piece` first appears, or `-1` if it's not there |
| `piece in text` | `True` if `piece` is somewhere in `text` |

Two more answer yes/no questions: `"2024".isdigit()` is `True` (all digits 0–9), and `"abc123".isalnum()` is `True` (all letters or digits; "alnum" is short for alphabet + number).

## Split a sentence into words, and join them back

**`split()`** cuts a string into words wherever there are spaces. **`join()`** glues them back together, with the string before the dot in between.

````python
sentence = "fly me   to the moon"
words = sentence.split()
print(words)
print(words[-1])
print(len(words))
print(" ".join(words))
print("-".join(words))
````

```
['fly', 'me', 'to', 'the', 'moon']
moon
5
fly me to the moon
fly-me-to-the-moon
```

- The thing in square brackets is a **list**: many values kept in order. Lists are the next notebook. The same index tricks work on it: `words[-1]` is the last word.
- The extra spaces between "me" and "to" disappeared. `split()` with nothing in the brackets handles any number of spaces.
- You can split on something else too: `"a,b,c".split(",")` gives `['a', 'b', 'c']`.

## You can't change a string, only make a new one

You can't swap one bead on a necklace. Run this and read the last line of the error:

````python
word = "python"
word[0] = "j"
````

```
TypeError: 'str' object does not support item assignment
```

Strings can't be changed in place. Instead, build a **new** string from pieces:

````python
word = "python"
new_word = "j" + word[1:]
print(new_word)
print(word)
````

```
jython
python
```

The same is true for methods. `upper()`, `replace()` and the others **give you a new string**. The old one stays as it was. **Predict first:** what do the two `print`s show?

````python
s = "hello"
s.upper()
print(s)
s = s.upper()
print(s)
````

```
hello
HELLO
```

The first `s.upper()` made `"HELLO"` and then threw it away, because nothing kept it. There's no error, which makes this bug easy to miss. To keep it, put it back in the box: `s = s.upper()`.

## Numbers to text and back: `str()` and `int()`

- **`str(x)`** turns a number into text.
- **`int(x)`** turns text made of digits into a number.

Text has tools that numbers don't. You can't reverse the number `123`, but you can reverse the text `"123"`:

````python
n = 123
print(str(n)[::-1])
print(int("42") + 1)
print(len(str(98765)))
````

```
321
43
5
```

`len(str(98765))` is a quick way to count the digits in a number.

## f-strings: put values inside text

An **f-string** has an `f` before the quotes. Anything inside `{ }` is worked out and dropped into the text.

````python
name = "Tanvi"
age = 20
print(f"{name} is {age} years old")
print(f"{name} has {len(name)} letters, backwards it's {name[::-1]}")
````

```
Tanvi is 20 years old
Tanvi has 5 letters, backwards it's ivnaT
```

## Common mistakes

**An index that doesn't exist.** `"python"` has indexes 0 to 5. Run this:

````python
word = "python"
print(word[6])
````

```
IndexError: string index out of range
```

"There's no letter at that position." The last index is `len(word) - 1`.

(Slices are kinder: `word[2:100]` gives `thon` with no error. Only a single index can be out of range.)

**Adding text and a number with `+`:**

````python
print("Age: " + 20)
````

```
TypeError: can only concatenate str (not "int") to str
```

**Concatenate** means "join together". Python won't join text and a number. Use `"Age: " + str(20)`, or an f-string: `f"Age: {20}"`.
