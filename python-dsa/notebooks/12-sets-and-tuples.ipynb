# 12. Sets and tuples

After this notebook you'll be able to keep a collection with no repeats, check "have I seen this?" fast, compare two groups, and keep two values together as a pair.

Two small tools. Both are simpler than dictionaries, and both show up a lot in LeetCode answers.

## Part 1: sets

### A guest list

Think of the guest list at the door of a wedding. Each name is on it once: writing "Tanvi" twice doesn't make two Tanvis. Nobody cares what order the names were written in. The only question the guard asks is: **"Is this name on the list?"** And the answer has to come fast.

That's a **set**: a collection where

- every item appears **only once** (no duplicates),
- there's **no order** (no index, no "first item"),
- checking `in` is **fast**, like looking up a key in a dictionary.

A set is a dictionary with only keys, and no values.

### Making a set: repeats disappear

````python
nums = [3, 1, 3, 2, 1, 3]
unique = set(nums)
print(unique)
print(len(nums))
print(len(unique))
````

```
{1, 2, 3}
6
3
```

`len(set(nums))` is the quick way to count **how many different** things there are. If it's smaller than `len(nums)`, something was repeated.

A set is written with curly brackets, like a dictionary, but with single items instead of `key: value` pairs.

A string becomes a set of its letters. Predict the number:

````python
print(len("hello"))
print(len(set("hello")))
````

```
5
4
```

`h`, `e`, `l`, `o`: the two `l`s count once.

A set of letters may print in a different order each time you run it (for example `{'o', 'h', 'l', 'e'}`). That's normal: a set has no order. Don't rely on it.

### Adding, removing, checking

Start an empty set with **`set()`**, then use **`add`** and **`remove`**:

````python
guests = set()
guests.add("Tanvi")
guests.add("Riya")
guests.add("Tanvi")
print(len(guests))
print("Riya" in guests)
print("Aman" in guests)
guests.remove("Riya")
print("Riya" in guests)
````

```
2
True
False
False
```

Adding `"Tanvi"` a second time changed nothing. She was already on the list.

**Why a set and not a list?** Both can answer `x in ...`. A list checks its items one by one, so a long list is slow to search. A set jumps straight to the answer, like a dictionary. On LeetCode, with big inputs, that difference can decide whether your answer passes or runs out of time.

### "Have I seen this before?" with a set

In the last notebook you used a `seen` **dictionary** to remember where you saw each number. If you only need **whether** you saw it (not where), a set is enough:

````python
seen = set()
for n in [4, 7, 2, 7, 9]:
    print("n =", n, ", seen =", seen)
    if n in seen:
        print("repeat:", n)
        break
    seen.add(n)
````

```
n = 4 , seen = set()
n = 7 , seen = {4}
n = 2 , seen = {4, 7}
n = 7 , seen = {2, 4, 7}
repeat: 7
```

Same rule as before: **check first, then add**.

An empty set prints as `set()`, not `{}`. You'll see why in "Common mistakes".

### Comparing two sets: `&`, `|`, `-`

You and Vikash each have a set of friends. Who's in both? Who's in either? Who's only yours?

````python
a = {1, 2, 3, 4}
b = {3, 4, 5}
print(a & b)
print(a | b)
print(a - b)
print(b - a)
````

```
{3, 4}
{1, 2, 3, 4, 5}
{1, 2}
{5}
```

```
 a = {1, 2, 3, 4}        b = {3, 4, 5}

   only in a    in both    only in b
     1  2        3  4         5
```

- **`a & b`** ("and"): items in **both**.
- **`a | b`** ("or"): items in **either** one.
- **`a - b`**: items in `a` that are **not** in `b`. Order matters: `b - a` is different.

If a problem wants a list back, turn the set into one with `list(...)`, or `sorted(...)` to get it in order: `sorted(a & b)` gives `[3, 4]`.

## Part 2: tuples

### A fixed pair

Think of a date of birth: day, month, year. Or a seat in a cinema: row and seat number. They always go together, and you don't change one half on its own.

A **tuple** is a small group of values kept together that **can't be changed** once made. You write it with round brackets:

````python
seat = (7, 12)
print(seat[0])
print(seat[1])
print(len(seat))
````

```
7
12
2
```

Indexing works like a list. Changing doesn't. Run this; it breaks on purpose:

````python
seat[0] = 8
````

```
TypeError: 'tuple' object does not support item assignment
```

Same as strings. If you need a different one, make a new tuple: `seat = (8, 12)`.

### Unpacking: one name per value

You can take a tuple apart into separate variables in one line. This is called **unpacking**:

````python
seat = (7, 12)
row, number = seat
print("row", row, "seat", number)
````

```
row 7 seat 12
```

The first value goes into the first name, the second into the second.

You've done this already without knowing it. `enumerate` and `.items()` hand you a tuple each time round, and `for i, n in ...` unpacks it:

````python
for pair in enumerate(["tea", "coffee"]):
    print(pair)

for pair in {"tea": 10, "coffee": 30}.items():
    print(pair)
````

```
(0, 'tea')
(1, 'coffee')
('tea', 10)
('coffee', 30)
```

### Swapping two values

Predict what this prints:

````python
a = 1
b = 2
a, b = b, a
print(a, b)

letters = ["x", "y", "z"]
letters[0], letters[2] = letters[2], letters[0]
print(letters)
````

```
2 1
['z', 'y', 'x']
```

Python first makes the tuple on the right, `(b, a)`, which is `(2, 1)`, and then unpacks it into `a` and `b`. The same trick swaps two items in a list. You'll use it a lot in the two-pointers notebook.

### Returning two values from a function

A function returns one thing. But that one thing can be a tuple, so you can hand back two answers at once:

````python
def lowest_and_highest(marks):
    return min(marks), max(marks)

low, high = lowest_and_highest([67, 92, 45, 80])
print(low)
print(high)
print(lowest_and_highest([67, 92, 45, 80]))
````

```
45
92
(45, 92)
```

`return min(marks), max(marks)` returns the tuple `(45, 92)`, and the caller unpacks it.

### Tuples as dictionary keys

A list can't be a dictionary key (or go in a set), because it can change. A tuple can't change, so it **can** be a key. That's handy for positions like (row, column):

````python
bookings = {}
bookings[(7, 12)] = "Tanvi"
bookings[(7, 13)] = "Riya"
print(bookings[(7, 12)])
print((7, 13) in bookings)
print((1, 1) in bookings)
````

```
Tanvi
True
False
```

You'll use this in the grids notebook, to remember which cells you've visited.

## Common mistakes

**`{}` is an empty dictionary, not an empty set.**

````python
print(type({}))
print(type(set()))
````

```
<class 'dict'>
<class 'set'>
```

For an empty set, always write `set()`.

**Using an index on a set.** Run this:

````python
s = {1, 2, 3}
print(s[0])
````

```
TypeError: 'set' object is not subscriptable
```

A set has no order, so there's no "item 0". If you need positions, use a list.

**Removing something that isn't there.** `guests.remove("Aman")` when Aman isn't on the list gives `KeyError: 'Aman'`. Check with `in` first.

**Unpacking into the wrong number of names.** `x, y = (1, 2, 3)` gives `ValueError: too many values to unpack (expected 2)`. You need one name per value.
