# 3. If and else

After this notebook you'll be able to make your code choose what to do, with `if`, `elif` and `else`.

"If it's raining, take an umbrella. Otherwise, take sunglasses." You ask a yes-or-no question, and the answer decides what you do. Code works the same way: it asks a question, gets `True` or `False`, and picks a path.

## Questions with True or False answers

A **comparison** is a question whose answer is a bool (`True` or `False`):

| Write | Means | Example | Answer |
| --- | --- | --- | --- |
| `==` | is equal to | `4 == 4` | `True` |
| `!=` | is not equal to | `4 != 4` | `False` |
| `<` | is less than | `2 < 5` | `True` |
| `>` | is more than | `2 > 5` | `False` |
| `<=` | is less than or equal to | `3 <= 3` | `True` |
| `>=` | is more than or equal to | `2 >= 5` | `False` |

### `=` and `==` are different

- **`=`** puts a value in a box: `age = 20`.
- **`==`** asks a question: "is age equal to 20?"

````python
age = 20
print(age == 20)
print(age == 18)
print("cat" == "Cat")
````

```
True
False
False
```

Capitals count: `"cat"` and `"Cat"` are different text.

## Joining questions: `and`, `or`, `not`

- **`and`**: `True` only if **both** sides are true.
- **`or`**: `True` if **at least one** side is true.
- **`not`**: flips the answer.

**Predict, then run:** with `age = 20`, what will each line give?

````python
age = 20
print(age >= 18 and age < 60)
print(age < 10 or age > 60)
print(not age > 18)
````

```
True
False
False
```

The last one: `age > 18` is `True`, and `not` flips it to `False`.

## `if`, `else` and `elif`

### `if`: do this only when it's true

````python
age = 20

if age >= 18:
    print("You can vote")

print("Done")
````

```
You can vote
Done
```

The shape matters:

```
if age >= 18:               <- the question, then a colon
    print("You can vote")   <- 4 spaces in: runs only if the answer is True
print("Done")               <- back at the left edge: runs every time
```

The pushed-in lines are called **indentation**. They belong to the `if`.

**Try it:** change `age` to `15` and run the cell again. Only `Done` prints.

### `else`: otherwise, do this

Exactly one of the two parts runs:

````python
number = 7

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
````

```
7 is odd
```

### `elif`: more than two paths

`elif` is short for "else if":

````python
marks = 72

if marks >= 90:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Try again")
````

```
Grade B
```

Python checks from the top and stops at the **first** true one:

| Question | Answer | What happens |
| --- | --- | --- |
| `marks >= 90` | `False` | skip |
| `marks >= 70` | `True` | print "Grade B", skip the rest |
| `marks >= 50` | never asked | |
| `else` | never reached | |

`72` is also `>= 50`, but Python already stopped. So the order of the questions matters.

**Try it:** set `marks` to `95`, then `50`, then `10`. Predict each grade first.

## Common mistakes

**Forgetting to indent.** Run it:

````python
age = 20
if age >= 18:
print("adult")
````

```
IndentationError: expected an indented block after 'if' statement on line 2
```

Python expected the line after the `if` to be pushed in. Add 4 spaces before `print` (the Tab key does it).

**Using `=` instead of `==`.** `if age = 18:` gives `SyntaxError: invalid syntax. Maybe you meant '==' or ':=' instead of '='?` Use `==`.

**Forgetting the colon.** `if age >= 18` with no `:` gives `SyntaxError: expected ':'`.
