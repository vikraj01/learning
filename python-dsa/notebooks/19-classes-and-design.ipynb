## A blank form and filled-in forms

Think of a college admission form. The blank form says what every student must fill in: name, age. It isn't about any one student. Then each student fills in their own copy: one copy says "Tanvi, 20", another says "Riya, 19".

In Python:

- A **class** is the blank form. It says what every copy will have.
- An **object** is one filled-in copy. You can make as many as you like from one class.
- The things written on a copy (name, age) are its **attributes**.

## Your first class

Run this.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

a = Student("Tanvi", 20)
b = Student("Riya", 19)
print(a.name, a.age)
print(b.name, b.age)
````

```
Tanvi 20
Riya 19
```

Line by line:

1. `class Student:` starts the blank form, named `Student`. Class names start with a capital letter, by habit.
2. `def __init__(self, name, age):` is a function inside the class. A function inside a class is called a **method**. This one has a special name: `__init__` (two underscores on each side, short for "initialise", meaning "set up"). Python runs it **by itself** every time you make a new object. Its job is to fill in the form.
3. `self.name = name` means "on **this** copy, write `name` in the name box".
4. `Student("Tanvi", 20)` makes a new object. You call the class like a function, and you get a filled-in copy back.
5. `a.name` reads an attribute. The dot means "the `name` of `a`".

## What is `self`? Watch it happen

`self` is the copy being filled in **right now**. You never pass it yourself. Python makes a fresh, empty copy, hands it to `__init__` as `self`, and gives it back to you when `__init__` is done.

This version of `Student` prints what's happening inside `__init__`. Run it.

````python
class Student:
    def __init__(self, name, age):
        print("  filling in a new copy:", name, age)
        self.name = name
        self.age = age
        print("  done, this copy now says:", self.name, self.age)

print("making a")
a = Student("Tanvi", 20)
print("making b")
b = Student("Riya", 19)
print(a.name, "and", b.name)
````

```
making a
  filling in a new copy: Tanvi 20
  done, this copy now says: Tanvi 20
making b
  filling in a new copy: Riya 19
  done, this copy now says: Riya 19
Tanvi and Riya
```

`__init__` ran twice, once per object, and each time `self` was a **different** copy:

| Line you wrote | `self` inside `__init__` is | `self.name` | `self.age` | Then it's called |
| --- | --- | --- | --- | --- |
| `a = Student("Tanvi", 20)` | a fresh empty copy | `"Tanvi"` | `20` | `a` |
| `b = Student("Riya", 19)` | another fresh empty copy | `"Riya"` | `19` | `b` |

So the same line `self.name = name` wrote "Tanvi" on one form and "Riya" on another. `self` means "whichever copy we're working on".

## Methods: functions that belong to the object

A class can have more methods than `__init__`. Every method gets `self` first, so it can read and change **that** object's attributes.

````python
class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def describe(self):
        return f"{self.name} is {self.age} years old"

    def birthday(self):
        self.age += 1

tom = Cat("Tom", 3)
kitty = Cat("Kitty", 5)
print(tom.describe())
tom.birthday()
tom.birthday()
print(tom.describe())
print(kitty.describe())
````

```
Tom is 3 years old
Tom is 5 years old
Kitty is 5 years old
```

When you write `tom.birthday()`, Python passes `tom` in as `self`. So inside `birthday`, `self.age += 1` means `tom.age += 1`. Kitty's age stays the same.

Here's a way to see it. `tom.describe()` is a short way of writing `Cat.describe(tom)`: "use the `describe` from the `Cat` form, with `tom` as `self`". Run it:

````python
print(tom.describe())
print(Cat.describe(tom))
print(Cat.describe(kitty))
````

```
Tom is 5 years old
Tom is 5 years old
Kitty is 5 years old
```

Nobody writes `Cat.describe(tom)` in real code. But it shows where `self` comes from: it's the thing before the dot.

That's also why `class Solution` methods start with `self`. LeetCode makes one `Solution()` object and calls your method on it. In normal problems you don't use `self` at all.

## Default values

A parameter can have a **default value**: what it gets if the caller leaves it out. This works in any `def`, including `__init__`.

````python
class Cup:
    def __init__(self, drink, size="small"):
        self.drink = drink
        self.size = size

c1 = Cup("chai")
c2 = Cup("coffee", "large")
print(c1.drink, c1.size)
print(c2.drink, c2.size)
````

```
chai small
coffee large
```

`Cup("chai")` left out `size`, so it got `"small"`.

## `None`: "nothing there"

You met `None` in notebook 04: what a function gives back when it has no `return`. With classes it gets a new job. An attribute that holds `None` means **"there's nothing here"**: no best friend yet, no next item, empty.

`None` counts as `False` in an `if`, the same way an empty list does. `is None` is the usual way to check for it.

````python
class Person:
    def __init__(self, name, best_friend=None):
        self.name = name
        self.best_friend = best_friend

riya = Person("Riya")
tanvi = Person("Tanvi", riya)

print(riya.best_friend)
print(riya.best_friend is None)
print(tanvi.best_friend.name)
if not riya.best_friend:
    print("Riya has no best friend saved yet")
````

```
None
True
Riya
Riya has no best friend saved yet
```

Look at `tanvi.best_friend.name`. `tanvi.best_friend` is a whole `Person` object (Riya), so you can put another dot after it and read *her* name. An attribute can hold another object. The next notebook is built on exactly this idea.

## A class that remembers: "design" problems

Some LeetCode problems don't ask for one function. They ask you to **design a class**. LeetCode makes one object and then calls its methods many times, and your object has to **remember** things between calls.

Think of a scorer at a cricket match with a notebook. When the match starts, they open a fresh page (`__init__`). Every ball, they write something down (`add`). When someone asks "what's the best score?", they read their notebook (`best`). The notebook is `self`.

````python
class ScoreBoard:
    def __init__(self):
        self.scores = []

    def add(self, score):
        self.scores.append(score)

    def best(self):
        return max(self.scores)

    def count(self):
        return len(self.scores)

board = ScoreBoard()
board.add(40)
board.add(75)
board.add(62)
print(board.best())
print(board.count())
````

```
75
3
```

Step by step, this is what's on the object after each line:

| Line | `board.scores` afterwards | Gives back |
| --- | --- | --- |
| `board = ScoreBoard()` | `[]` | the new object |
| `board.add(40)` | `[40]` | `None` |
| `board.add(75)` | `[40, 75]` | `None` |
| `board.add(62)` | `[40, 75, 62]` | `None` |
| `board.best()` | `[40, 75, 62]` | `75` |
| `board.count()` | `[40, 75, 62]` | `3` |

- `__init__` sets things up: an empty list, stored on the object as `self.scores`.
- Every other method reaches that same list through `self.scores`. That's how `best` sees what `add` stored.

### Why `self.scores` and not `scores`?

A plain variable inside a method is forgotten when the method ends, like any variable inside a function. Only what you put on `self` stays. This version forgets. Run it and read the last line of the error:

````python
class ForgetfulBoard:
    def __init__(self):
        scores = []          # no self. : forgotten when __init__ ends

    def add(self, score):
        scores.append(score)

fb = ForgetfulBoard()
fb.add(40)
````

```
NameError: name 'scores' is not defined
```

`scores` only lived inside `__init__`. When `add` runs, there's no `scores` anywhere. Put `self.` in front, in both places, and it works.

### How LeetCode writes a design problem

A design problem's examples come as two lists: the method names, and the arguments for each call. Here's problem 1603, Design Parking System:

```

[[1, 1, 0], [1], [2], [3], [1]]
Output
[null, true, true, false, false]
```

Read them **in columns**, one call each:

| Call | Gives back |
| --- | --- |
| `ParkingSystem(1, 1, 0)` | `null` (it makes the object) |
| `addCar(1)` | `true` |
| `addCar(2)` | `true` |
| `addCar(3)` | `false` |
| `addCar(1)` | `false` |

(`null` is how LeetCode writes `None`, and `true`/`false` are `True`/`False`.) On your laptop you do the same calls yourself, one per line: `p = ParkingSystem(1, 1, 0)`, then `p.addCar(1)`, and so on. That's Task 6 below.

## Common mistakes

**Forgetting `self` as the first parameter.** Run this:

````python
class Box:
    def add(x):
        print(x)

Box().add(5)
````

```
TypeError: Box.add() takes 1 positional argument but 2 were given
```

Python always passes the object as the first argument, so `add` gets two things (the object and `5`) but only has room for one. Every method starts with `self`.

**Forgetting the brackets when you call a method.**

````python
board = ScoreBoard()
board.add(10)
print(board.best)
print(board.best())
````

The first line prints something like `<bound method ScoreBoard.best of <__main__.ScoreBoard object at 0x7f...>>`. That's the method itself, not its answer. Without `()`, nothing runs. The second line, with brackets, prints `10`.
